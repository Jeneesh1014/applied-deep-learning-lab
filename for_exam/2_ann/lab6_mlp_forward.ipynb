## 🔹 Task 1: Bias Neuron


1. Explain the role of a bias neuron in artificial neural networks

A bias neuron is a special neuron that always has a constant value, usually 1. It is added to the weighted sum of the inputs before the activation function is applied. The main role of the bias neuron is to help the neural network adjust or shift the output of a neuron. This allows the network to learn patterns more effectively and make better predictions. Bias works similarly to the intercept term in a linear equation because it helps control where the activation starts. Without bias, the neuron output would be more restricted and less flexible.


2. Describe the effects of using a bias neuron

Using a bias neuron improves the flexibility and learning ability of a neural network. It allows neurons to activate even when the input values are small or zero. Bias also helps shift the decision boundary, so the network can represent more complex patterns in data. Without a bias neuron, the output is often forced to pass through the origin, which can reduce accuracy and make learning difficult. Therefore, bias neurons help the network train better and produce more accurate results.

Q1. How does neuron output depend on inputs?

The output of a neuron depends on the input values because the inputs are multiplied by weights and summed together. Different input values produce different weighted sums, which change the final output after applying the activation function.


Q2. How does neuron output depend on weights?

Weights determine the importance of each input in a neuron. Larger weights give more influence to an input, while smaller weights reduce its effect. Negative weights can suppress the contribution of an input.


Q4. What is the mathematical representation of a neuron?

The output of a neuron is calculated using the weighted sum of inputs and bias followed by an activation function:

y=f(∑wixi + b)

where:

xi are inputs

wi are weights

b is bias

f is activation function


## 🔹 Task 3: Manual Forward Pass Calculation


# Forward Propagation in a Multi-Layer Neural Network

---

## Goal

Calculate the following using manual forward propagation:

- Hidden neuron outputs ($H_1$, $H_2$)
- Final predicted output
- Prediction error

---

## Given Values

### Inputs

$$x_1 = 2, \qquad x_2 = 1$$

---

### Hidden Layer Weights

**Hidden Neuron $H_1$:**

$$w_{11} = 0.5, \qquad w_{12} = 0.3, \qquad b_1 = 0.1$$

**Hidden Neuron $H_2$:**

$$w_{21} = 0.4, \qquad w_{22} = 0.2, \qquad b_2 = 0.2$$

---

### Output Layer Weights

$$v_1 = 0.6, \qquad v_2 = 0.5, \qquad b_{out} = 0.3$$

---

### Activation Function — ReLU

$$\text{ReLU}(z) = \max(0,\ z)$$

ReLU outputs the input directly if it is positive, and outputs 0 if it is negative.

---

### Target Output

$$y = 2.0$$

---

## General Formula for Each Neuron

For any neuron, the computation has two steps:

**Step 1 — Weighted Sum (pre-activation):**

$$z = \sum_{i} (x_i \times w_i) + b$$

**Step 2 — Apply Activation Function (post-activation):**

$$\text{output} = \text{ReLU}(z) = \max(0,\ z)$$

---

## Step 1 — Calculate Hidden Neuron $H_1$

### Weighted Sum

$$z_1 = (x_1 \times w_{11}) + (x_2 \times w_{12}) + b_1$$

$$z_1 = (2 \times 0.5) + (1 \times 0.3) + 0.1$$

$$z_1 = 1.0 + 0.3 + 0.1$$

$$z_1 = 1.4$$

### Apply ReLU

$$H_1 = \text{ReLU}(z_1) = \max(0,\ 1.4) = \boxed{1.4}$$

---

## Step 2 — Calculate Hidden Neuron $H_2$

### Weighted Sum

$$z_2 = (x_1 \times w_{21}) + (x_2 \times w_{22}) + b_2$$

$$z_2 = (2 \times 0.4) + (1 \times 0.2) + 0.2$$

$$z_2 = 0.8 + 0.2 + 0.2$$

$$z_2 = 1.2$$

### Apply ReLU

$$H_2 = \text{ReLU}(z_2) = \max(0,\ 1.2) = \boxed{1.2}$$

---

## Step 3 — Calculate Final Output

### Weighted Sum

$$z_{out} = (H_1 \times v_1) + (H_2 \times v_2) + b_{out}$$

$$z_{out} = (1.4 \times 0.6) + (1.2 \times 0.5) + 0.3$$

$$z_{out} = 0.84 + 0.60 + 0.30$$

$$z_{out} = 1.74$$

### Apply ReLU

$$\hat{y} = \text{ReLU}(z_{out}) = \max(0,\ 1.74) = \boxed{1.74}$$

---

## Step 4 — Calculate Prediction Error

$$\text{Prediction Error} = y - \hat{y}$$

$$\text{Prediction Error} = 2.0 - 1.74$$

$$\text{Prediction Error} = \boxed{0.26}$$

---

## Full Forward Propagation Flow

```
Inputs
  x₁ = 2,  x₂ = 1
        │
        ▼
Hidden Layer
  z₁ = (2×0.5) + (1×0.3) + 0.1 = 1.4  →  H₁ = ReLU(1.4) = 1.40
  z₂ = (2×0.4) + (1×0.2) + 0.2 = 1.2  →  H₂ = ReLU(1.2) = 1.20
        │
        ▼
Output Layer
  z_out = (1.4×0.6) + (1.2×0.5) + 0.3 = 1.74
  ŷ = ReLU(1.74) = 1.74
        │
        ▼
Target:  y = 2.00
        │
        ▼
Error = y − ŷ = 2.00 − 1.74 = 0.26
```
