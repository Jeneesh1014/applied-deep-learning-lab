# Feature Scaling & Standardization — Complete Study Notes

---

## 1. What is Feature Scaling?

Feature Scaling is a **preprocessing technique** used in Machine Learning to bring all features (columns) to a **similar numerical range**.

### Why do we need it?

Imagine a dataset with two features:

| Person | Age | Salary (€) |
| ------ | --- | ---------- |
| A      | 25  | 50,000     |
| B      | 30  | 80,000     |
| C      | 22  | 30,000     |

- Age ranges from **22 to 30** (small numbers)
- Salary ranges from **30,000 to 80,000** (very large numbers)

When a machine learning model calculates distances or weights, **Salary will dominate completely** just because its numbers are bigger — not because it is actually more important. This is wrong and leads to poor model performance.

> **Solution:** Scale all features to the same range so every feature gets a fair chance.

---

### Which algorithms NEED feature scaling?

| Algorithm                     | Needs Scaling? | Why?                                  |
| ----------------------------- | -------------- | ------------------------------------- |
| K-Nearest Neighbors (KNN)     | ✅ Yes         | Uses distance — large values dominate |
| Support Vector Machine (SVM)  | ✅ Yes         | Uses distance — large values dominate |
| Linear / Logistic Regression  | ✅ Yes         | Gradient descent converges faster     |
| Neural Networks               | ✅ Yes         | Stability during training             |
| Decision Tree / Random Forest | ❌ No          | Uses rules, not distances             |
| Naive Bayes                   | ❌ No          | Uses probabilities, not distances     |

## 2. Standardization

**Standardization** is the process of transforming features so they all operate on the same scale.

There are **two main methods:**

---

## Method 1 — Min-Max Normalization

### Definition

Min-Max Normalization **rescales** every value so that it fits within a **fixed range of [0, 1]**, while keeping the same distribution shape (the pattern of data does not change).

### Formula

$$x' = \frac{x - x_{\min}}{x_{\max} - x_{\min}}$$

### What each symbol means

| Symbol     | Meaning                                       |
| ---------- | --------------------------------------------- |
| $x$        | The original value you want to scale          |
| $x_{\min}$ | The smallest value in that feature/column     |
| $x_{\max}$ | The largest value in that feature/column      |
| $x'$       | The new scaled value (always between 0 and 1) |

### Intuition (How to think about it)

- The **minimum value** always becomes **0**
- The **maximum value** always becomes **1**
- All other values fall **proportionally between 0 and 1**

### Worked Example

Suppose we have the following **Age** column:

| Person | Age |
| ------ | --- |
| A      | 20  |
| B      | 30  |
| C      | 40  |
| D      | 50  |

- $x_{\min} = 20$, $x_{\max} = 50$

Now apply the formula to each:

**Person A (Age = 20):**
$$x' = \frac{20 - 20}{50 - 20} = \frac{0}{30} = 0.0$$

**Person B (Age = 30):**
$$x' = \frac{30 - 20}{50 - 20} = \frac{10}{30} = 0.33$$

**Person C (Age = 40):**
$$x' = \frac{40 - 20}{50 - 20} = \frac{20}{30} = 0.67$$

**Person D (Age = 50):**
$$x' = \frac{50 - 20}{50 - 20} = \frac{30}{30} = 1.0$$

**Result:**

| Person | Original Age | Scaled Age (Min-Max) |
| ------ | ------------ | -------------------- |
| A      | 20           | 0.00                 |
| B      | 30           | 0.33                 |
| C      | 40           | 0.67                 |
| D      | 50           | 1.00                 |

> Notice: the distribution **shape is preserved** — the spacing between values remains proportional.

### Advantages

- Simple and easy to understand
- Output is always in a known, fixed range [0, 1]
- Good when you need bounded inputs (e.g., neural network activation functions)

### Disadvantages

- **Very sensitive to outliers** — if one value is extremely large or small, it compresses all other values into a tiny range

**Example of outlier problem:**

Ages: 20, 30, 40, **500** (one outlier)

$$x'_{30} = \frac{30 - 20}{500 - 20} = \frac{10}{480} \approx 0.02$$

Now 30 and 40 both get values close to 0 — the outlier destroys the scaling.

### When to use Min-Max Normalization

→ Use when:

- A **fixed numeric range is mandatory** (e.g., image pixel values must be 0–1)
- Your data has **no extreme outliers**
- You are using algorithms like **Neural Networks** or **KNN**

## Method 2 — Z-Score Normalization (Standardization)

### Definition

Z-Score Normalization **centers** the data around **0** and scales it so the **standard deviation becomes 1**. Instead of asking "where does this value fall in the [0,1] range?", it asks: **"How many standard deviations away from the mean is this value?"**

### Formula

$$x' = \frac{x - \mu}{\sigma}$$

### What each symbol means

| Symbol           | Meaning                                           |
| ---------------- | ------------------------------------------------- |
| $x$              | The original value you want to scale              |
| $\mu$ (mu)       | The **mean** (average) of the feature/column      |
| $\sigma$ (sigma) | The **standard deviation** of the feature/column  |
| $x'$             | The new standardized value (called a **z-score**) |

### Intuition (How to think about it)

- If $x' = 0$ → the value is **exactly at the mean**
- If $x' = 1$ → the value is **1 standard deviation above** the mean
- If $x' = -1$ → the value is **1 standard deviation below** the mean
- If $x' = 2$ → the value is **far above** the mean (possible outlier)

> There is **no fixed range** — values can go below 0 or above 1. The focus is on **relative position** from the center.

### Worked Example

Same **Age** column:

| Person | Age |
| ------ | --- |
| A      | 20  |
| B      | 30  |
| C      | 40  |
| D      | 50  |

**Step 1: Calculate the mean ($\mu$)**

$$\mu = \frac{20 + 30 + 40 + 50}{4} = \frac{140}{4} = 35$$

**Step 2: Calculate the standard deviation ($\sigma$)**

$$\sigma = \sqrt{\frac{(20-35)^2 + (30-35)^2 + (40-35)^2 + (50-35)^2}{4}}$$
$$= \sqrt{\frac{225 + 25 + 25 + 225}{4}} = \sqrt{\frac{500}{4}} = \sqrt{125} \approx 11.18$$

**Step 3: Apply the formula to each value**

**Person A (Age = 20):**
$$x' = \frac{20 - 35}{11.18} = \frac{-15}{11.18} \approx -1.34$$

**Person B (Age = 30):**
$$x' = \frac{30 - 35}{11.18} = \frac{-5}{11.18} \approx -0.45$$

**Person C (Age = 40):**
$$x' = \frac{40 - 35}{11.18} = \frac{5}{11.18} \approx +0.45$$

**Person D (Age = 50):**
$$x' = \frac{50 - 35}{11.18} = \frac{15}{11.18} \approx +1.34$$

**Result:**

| Person | Original Age | Z-Score |
| ------ | ------------ | ------- |
| A      | 20           | -1.34   |
| B      | 30           | -0.45   |
| C      | 40           | +0.45   |
| D      | 50           | +1.34   |

> Notice: values are now **centered around 0**, and the mean of z-scores = 0.

### Advantages

- **Less sensitive to outliers** than Min-Max (outliers just get a large z-score, but do not crush other values)
- Works well when the data follows a **normal (bell-curve) distribution**
- Preferred by most ML algorithms

### Disadvantages

- Output is **not bounded** — you cannot guarantee values stay in [0, 1]
- Slightly harder to interpret than Min-Max

### When to use Z-Score Normalization

→ Use when:

- Features have **different scales** and you don't need a fixed range
- Your data has **no extreme outliers** (but handles them better than Min-Max)
- You are using algorithms like **SVM, Logistic Regression, PCA, Neural Networks**

---

## 3. Side-by-Side Comparison

|                                   | Min-Max Normalization                               | Z-Score Normalization                   |
| --------------------------------- | --------------------------------------------------- | --------------------------------------- |
| **Formula**                       | $x' = \dfrac{x - x_{\min}}{x_{\max} - x_{\min}}$    | $x' = \dfrac{x - \mu}{\sigma}$          |
| **Output range**                  | Always [0, 1]                                       | Unbounded (centered at 0)               |
| **Mean after scaling**            | Not always 0                                        | Always **0**                            |
| **Std Dev after scaling**         | Not always 1                                        | Always **1**                            |
| **Sensitive to outliers?**        | ✅ Very sensitive                                   | ❌ Less sensitive                       |
| **Preserves distribution shape?** | Yes                                                 | Yes                                     |
| **Best for**                      | Fixed-range requirements (e.g. images, neural nets) | General ML algorithms, different scales |

## 4. Encoding Reference (Categorical Variables)

When your data has **categorical** (text) columns, you must convert them to numbers. The method depends on whether the categories have a natural order:

|                       | Ordinal Encoding              | One-Hot Encoding                                                                                                     |
| --------------------- | ----------------------------- | -------------------------------------------------------------------------------------------------------------------- |
| **Used when**         | Categories **have order**     | Categories **have no order**                                                                                         |
| **Example**           | Low → 1, Medium → 2, High → 3 | Red, Blue, Green → 3 separate 0/1 columns                                                                            |
| **Why not mix them?** | —                             | Using numbers for unordered categories implies false order (e.g., Red=1, Blue=2 suggests Blue > Red, which is wrong) |

## 5. Key Points to Remember for the Exam

> ✏️ **Definition of Standardization:**
> Standardization is a preprocessing step that rescales feature values to a common range so that no feature dominates the model due to its magnitude.

> ✏️ **Min-Max Formula:** $x' = \dfrac{x - x_{\min}}{x_{\max} - x_{\min}}$ → output always in [0, 1]

> ✏️ **Z-Score Formula:** $x' = \dfrac{x - \mu}{\sigma}$ → output centered at 0 with std dev = 1

> ✏️ **When to choose Min-Max:** Fixed range is required; no outliers

> ✏️ **When to choose Z-Score:** Features have different scales; better outlier handling

> ✏️ **Both methods preserve the shape of the original distribution** — they only change the scale, not the pattern.

> ✏️ **Algorithms that DO NOT need scaling:** Decision Trees, Random Forests, Naive Bayes (they use rules/probabilities, not distances)
