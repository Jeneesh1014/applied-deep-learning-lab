# AI, Machine Learning & Deep Learning — Complete Study Notes

---

## 1. The Big Picture — Nested Relationship

The most important thing to understand is that these are **not separate things** — they are **nested subsets** of each other:

```
┌─────────────────────────────────────────────┐
│                 ARTIFICIAL INTELLIGENCE      │
│   (any technique that lets machines "think") │
│                                              │
│   ┌──────────────────────────────────────┐  │
│   │         MACHINE LEARNING             │  │
│   │  (learns from data automatically)    │  │
│   │                                      │  │
│   │   ┌────────────────────────────┐     │  │
│   │   │   REPRESENTATION LEARNING  │     │  │
│   │   │  (learns features itself)  │     │  │
│   │   │                            │     │  │
│   │   │   ┌──────────────────┐     │     │  │
│   │   │   │  DEEP LEARNING   │     │     │  │
│   │   │   │  (many layers)   │     │     │  │
│   │   │   └──────────────────┘     │     │  │
│   │   └────────────────────────────┘     │  │
│   └──────────────────────────────────────┘  │
└─────────────────────────────────────────────┘
```

> **Key rule:** Every Deep Learning model IS a Machine Learning model. Every Machine Learning model IS an AI system. But NOT every AI system uses Machine Learning.

---

## 2. Artificial Intelligence (AI)

### Definition

Artificial Intelligence is the **broadest field** — it covers any technique or system that enables a machine to **simulate human intelligence**, such as reasoning, problem solving, understanding language, or making decisions.

### Key idea

AI does not have to _learn_ from data. It can be based on **hard-coded rules** written by humans.

### Examples of AI that is NOT Machine Learning

- **Knowledge Bases / Expert Systems** — a system where human experts manually write thousands of rules:
  - e.g., "IF patient has fever AND cough → THEN possible flu"
  - The machine just follows rules. It does not learn. It does not improve.
- **Rule-based chatbots** — scripted responses, no learning involved
- **Search algorithms** (like A\*) — follow logical steps defined by programmers

### Limitations of classical AI (rule-based)

- Someone must manually write every rule
- Cannot handle new situations not covered by the rules
- Does not scale to complex real-world problems (e.g., recognizing faces, understanding speech)

---

## 3. Machine Learning (ML)

### Definition

Machine Learning is a **subset of AI** where systems **automatically learn patterns from data** without being explicitly programmed with rules. The machine improves its performance as it sees more data.

### Key idea

Instead of writing rules manually, you give the machine **data + the correct answers**, and it **figures out the rules itself**.

### The fundamental shift from AI to ML

| Classical AI                | Machine Learning                   |
| --------------------------- | ---------------------------------- |
| Programmer writes the rules | Machine learns the rules from data |
| Rules are fixed             | Rules improve with more data       |
| Fails on unseen situations  | Generalizes to new data            |
| Example: Expert system      | Example: Logistic Regression       |

### How ML works — simple pipeline

```
Raw Data  →  [Hand-crafted Features]  →  ML Algorithm  →  Prediction
              (done by humans)
```

> **Critical point:** In classical ML, a **human expert** must manually decide which features (columns, measurements) to extract from the raw data. This is called **feature engineering**.

### Feature Engineering in Classical ML — explained

Suppose you want to build a model to detect spam emails.

A human expert must manually decide:

- Count of exclamation marks `!`
- Presence of words like "FREE", "WIN", "CLICK"
- Length of the email
- Sender domain

The human extracts these features → feeds them into the algorithm → algorithm learns which features matter most.

**Problem:** This requires deep domain expertise. For complex data like images or audio, it is extremely difficult for humans to design the right features.

### Examples of Classical ML algorithms

| Algorithm                    | Task                                  |
| ---------------------------- | ------------------------------------- |
| Logistic Regression          | Binary classification (spam/not spam) |
| Decision Trees               | Classification & regression           |
| Random Forest                | Classification & regression           |
| Support Vector Machine (SVM) | Classification                        |
| K-Nearest Neighbors (KNN)    | Classification                        |
| K-Means                      | Clustering (unsupervised)             |

## 4. Representation Learning

### Definition

Representation Learning is a **subset of Machine Learning** where the model does **not just learn to predict** — it also **learns what features to look at** (learns its own representation of the data).

### Key idea

Instead of a human telling the model "look at these features", the model discovers on its own **which features are useful**.

### Example — Shallow Autoencoders

An autoencoder is a type of neural network that:

1. Takes input data (e.g., an image)
2. **Compresses it** into a smaller representation (encoding)
3. **Reconstructs it** back to the original

In doing so, it learns a compact, meaningful representation of the data — without human guidance.

```
Input Image  →  [Encoder]  →  Compressed Representation  →  [Decoder]  →  Reconstructed Image
```

The compressed representation = the features the model learned by itself.

---

## 5. Deep Learning (DL)

### Definition

Deep Learning is a **subset of Representation Learning** that uses **artificial neural networks with many layers** (hence "deep"). Each layer automatically learns increasingly abstract features from the raw data.

### Key idea

Deep Learning takes representation learning further — instead of just learning one level of features, it learns **hierarchical features** layer by layer, starting from very simple to very complex.

### How Deep Learning processes data — Layer by Layer

Using image recognition as an example:

```
Raw Pixels
    ↓
Layer 1: Learns edges and lines
    ↓
Layer 2: Learns shapes and corners
    ↓
Layer 3: Learns object parts (eyes, wheels, doors)
    ↓
Layer 4: Learns full objects (face, car)
    ↓
Output: "This is a cat"
```

No human told it to look for edges or shapes — it discovered this hierarchy on its own.

### Why "Deep"?

"Deep" refers to the **number of layers** in the neural network:

- A network with 1–2 hidden layers → **Shallow network** (e.g., Shallow Autoencoder)
- A network with many hidden layers → **Deep network** (e.g., MLP with many layers, CNN, RNN)

### Examples of Deep Learning models

| Model       | Full Name                      | Common Use                        |
| ----------- | ------------------------------ | --------------------------------- |
| MLP         | Multi-Layer Perceptron         | General classification/regression |
| CNN         | Convolutional Neural Network   | Image recognition                 |
| RNN         | Recurrent Neural Network       | Sequence data, time series        |
| LSTM        | Long Short-Term Memory         | Language, speech                  |
| Transformer | —                              | NLP, ChatGPT, translation         |
| GAN         | Generative Adversarial Network | Image generation                  |

## 6. Feature Extraction: Classical ML vs Deep Learning

> This is one of the most important exam questions in this topic.

### Classical Machine Learning — Manual Feature Extraction

In classical ML, **a human expert manually designs and selects features** from raw data before feeding it to the model.

```
Raw Data
   ↓
[Human Expert analyses data]
[Human decides: "I will extract these features"]
   ↓
Feature Table (structured, hand-crafted)
   ↓
ML Algorithm (Logistic Regression, SVM, etc.)
   ↓
Prediction
```

**Characteristics:**

- Feature extraction is a **separate step** done by humans
- Requires **domain knowledge** (e.g., a medical expert to decide which symptoms matter)
- Works well when features are clear and interpretable
- Struggles with **complex, high-dimensional data** (images, audio, text)

**Example:**
To classify emails as spam:

- A human extracts: word counts, punctuation count, sender info, etc.
- These features go into Logistic Regression or SVM

### Deep Learning — Automatic Feature Extraction

In Deep Learning, **the model itself learns what features to extract** directly from raw data. There is no separate feature engineering step.

```
Raw Data (pixels, audio waves, raw text)
   ↓
[Deep Neural Network — Layer 1 extracts basic features]
[Layer 2 combines them into higher-level features]
[Layer 3 combines those into even more abstract features]
   ↓
Prediction
```

**Characteristics:**

- Feature extraction is **built into the model** — happens automatically
- Requires **no domain expertise** for feature design
- Needs **large amounts of data** to learn good features
- Needs **more compute power** (GPUs)
- Outperforms classical ML on complex unstructured data (images, audio, text)

**Example:**
To classify images of cats vs dogs:

- A CNN automatically learns: edges → shapes → animal parts → full animal
- No human defines what to look for

### Direct Comparison Table

| Aspect                          | Classical ML                            | Deep Learning                    |
| ------------------------------- | --------------------------------------- | -------------------------------- |
| **Feature extraction**          | Manual — done by human experts          | Automatic — learned by the model |
| **Domain knowledge needed**     | Yes — to engineer features              | No — for feature extraction      |
| **Raw input**                   | Structured, pre-processed features      | Raw data (pixels, audio, text)   |
| **Data requirement**            | Works with small datasets               | Needs large datasets             |
| **Compute requirement**         | Low — runs on CPU                       | High — needs GPU                 |
| **Interpretability**            | Higher — features are human-defined     | Lower — features are abstract    |
| **Performance on images/audio** | Poor (hard to hand-craft features)      | Excellent                        |
| **Performance on small data**   | Good                                    | Poor (overfits easily)           |
| **Examples**                    | SVM, Logistic Regression, Random Forest | CNN, RNN, MLP, Transformer       |

## 8. Key Exam Points to Remember

> ✏️ **Relationship:** Deep Learning ⊂ Representation Learning ⊂ Machine Learning ⊂ AI
> (each is a subset of the one outside it)

> ✏️ **AI without ML:** AI can exist without ML — e.g., expert systems, knowledge bases, rule-based systems.

> ✏️ **ML without DL:** ML can exist without Deep Learning — e.g., Logistic Regression, SVM, Decision Trees.

> ✏️ **Classical ML feature extraction:** Done **manually by humans** before training. Requires domain expertise. Called "feature engineering".

> ✏️ **Deep Learning feature extraction:** Done **automatically by the network** during training. Each layer learns increasingly abstract features from raw data.

> ✏️ **Why Deep Learning needs more data:** Because it must learn features from scratch — classical ML starts with human-designed features, so less data is needed.

> ✏️ **"Deep" means many layers** — more layers = more abstract features = better performance on complex data.

> ✏️ **Representation Learning** = the model learns _how to represent_ the data (what features to use), not just the final answer.
