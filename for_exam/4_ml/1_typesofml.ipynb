# Introduction

Machine Learning algorithms can be classified into four major types based on how they learn from data. These are:

1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning
4. Self-Supervised Learning

Each learning approach is designed to solve different types of problems depending on the availability of labeled data and the learning objective.

---

# 1. Supervised Learning

## Definition

Supervised Learning is a type of machine learning in which the model is trained using **labeled data**. Each training example contains both the input features and the correct output (label). The objective is to learn the relationship between inputs and outputs so that the model can accurately predict the output for new unseen data.

## Characteristics

- Uses labeled data.
- Learns from known input-output pairs.
- Used for Classification and Regression problems.
- Training data contains the correct answers.

## Applications

- Spam email detection
- House price prediction
- Disease diagnosis
- Image classification
- Credit risk prediction

## Example

A model is trained using emails labeled as **Spam** or **Not Spam**. After training, it can classify new emails automatically.

## Advantages

- High prediction accuracy when sufficient labeled data is available.
- Easy to evaluate using known target values.
- Suitable for many real-world prediction tasks.

## Disadvantages

- Requires a large amount of labeled data.
- Data labeling is expensive and time-consuming.

---

# 2. Unsupervised Learning

## Definition

Unsupervised Learning is a type of machine learning in which the model learns from **unlabeled data**. The objective is to discover hidden patterns, structures, or relationships within the data without knowing the correct outputs.

## Characteristics

- Uses unlabeled data.
- Finds similarities between data samples.
- Mainly used for clustering and pattern discovery.

## Applications

- Customer segmentation
- Market basket analysis
- Image segmentation
- Document clustering
- Anomaly detection

## Example

A retail company groups customers with similar purchasing behavior into different customer segments without predefined labels.

## Advantages

- Does not require labeled data.
- Can discover hidden patterns.
- Useful for exploratory data analysis.

## Disadvantages

- Results may be difficult to interpret.
- Accuracy is harder to evaluate because no correct labels exist.

---

# 3. Reinforcement Learning

## Definition

Reinforcement Learning is a type of machine learning in which an **agent learns by interacting with an environment**. The agent performs actions, receives rewards or penalties, and gradually learns the best strategy to maximize the total reward.

## Characteristics

- Learns through trial and error.
- Uses rewards and penalties as feedback.
- No labeled dataset is required.
- Focuses on sequential decision making.

## Main Components

- Agent
- Environment
- State
- Action
- Reward

## Applications

- Robotics
- Autonomous vehicles
- Game playing
- Industrial automation
- Resource management

## Example

A robot learns to navigate through a room. It receives positive rewards for reaching the destination and negative rewards when hitting obstacles. Over time, it learns the optimal path.

## Advantages

- Suitable for complex decision-making problems.
- Learns continuously from interaction.
- Can optimize long-term performance.

## Disadvantages

- Training may require many interactions.
- Computationally expensive.
- Designing an effective reward function can be challenging.

---

# 4. Self-Supervised Learning

## Definition

Self-Supervised Learning is a machine learning approach that uses **unlabeled data** but automatically generates labels (called pseudo-labels) from the data itself. The model first learns useful representations through a pretext task and then uses these learned representations for downstream tasks.

## Characteristics

- Uses unlabeled data.
- Automatically creates pseudo-labels.
- Reduces the need for manual data labeling.
- Widely used in modern Deep Learning.

## Applications

- Large Language Models (LLMs)
- Image representation learning
- Speech recognition
- Natural Language Processing
- Computer Vision

## Example

In language models such as GPT, some words are hidden or the next word is predicted. The hidden or next word serves as the training target, allowing the model to learn from unlabeled text.

## Advantages

- Reduces dependence on labeled datasets.
- Learns powerful feature representations.
- Improves performance on downstream tasks.

## Disadvantages

- Requires large amounts of data.
- Training can be computationally expensive.

---

# Conclusion

Machine Learning consists of four major learning approaches. Supervised Learning uses labeled data to make predictions, Unsupervised Learning discovers hidden structures in unlabeled data, Reinforcement Learning learns through rewards and penalties by interacting with an environment, and Self-Supervised Learning generates pseudo-labels from unlabeled data to learn meaningful representations. Each approach is suitable for different types of real-world AI applications.

---

# One-Line Revision

- **Supervised Learning:** Learns from labeled data.
- **Unsupervised Learning:** Finds patterns in unlabeled data.
- **Reinforcement Learning:** Learns by interacting with the environment using rewards.
- **Self-Supervised Learning:** Learns from unlabeled data by creating pseudo-labels automatically.
