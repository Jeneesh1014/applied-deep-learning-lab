# Graph Master Notes (Exam Revision)

**Applied Deep Learning / Data Visualization**

> These notes cover the most commonly used graphs in university exams. Focus on understanding **what each graph represents, when it should be used, and why it is the correct choice**.

---

# 1. Histogram

## Definition

A **Histogram** is a graph used to visualize the **distribution of numerical (quantitative) data**. It divides the data into intervals called **bins** and displays the number (frequency) of observations that fall within each interval.

Unlike a bar chart, a histogram does not display separate categories. Instead, it groups continuous numerical values into ranges.

---

## Purpose

A histogram helps answer questions such as:

- How is the data distributed?
- Which values occur most frequently?
- Is the data symmetric or skewed?
- Are there multiple peaks?
- Is the data spread out or concentrated?

---

## When to Use

Use a histogram when:

- The variable is **numerical**.
- You want to study the **distribution** of the data.
- The data contains many numerical values.
- You want to group values into intervals.

---

## Examples

### Example 1

Question:

> How are the body masses of penguins distributed?

Correct Graph:

**Histogram**

Reason:

Body mass is a numerical variable, and the objective is to study its distribution.

---

### Example 2

Question:

> How are students' exam marks distributed?

Correct Graph:

**Histogram**

Reason:

Exam marks are numerical values, and the histogram shows how frequently different score ranges occur.

---

## Important Characteristics

- Used for **numerical data**.
- Data is divided into **bins**.
- Bars **touch each other** because the intervals are continuous.
- Shows the **frequency distribution**.

---

# 2. Bar Chart

## Definition

A **Bar Chart** is a graph used to compare values or frequencies across **different categories**. Each bar represents one category, and the height (or length) of the bar represents its value or frequency.

Unlike a histogram, every bar represents a separate category rather than a numerical interval.

---

## Purpose

A bar chart helps answer questions such as:

- Which category has the highest value?
- Which category has the lowest value?
- How do different categories compare?

---

## When to Use

Use a bar chart when:

- The variable is **categorical**.
- You want to compare categories.
- You want to compare counts, averages, percentages, or totals for different groups.

---

## Examples

### Example 1

Question:

> How many penguins belong to each species?

Correct Graph:

**Bar Chart**

Reason:

Species is a categorical variable.

---

### Example 2

Question:

> Compare the average salary of different departments.

Correct Graph:

**Bar Chart**

Reason:

The departments are categories, while the average salary is the numerical value being compared.

---

## Important Characteristics

- Used for **categorical data**.
- Each bar represents one category.
- Bars have **spaces (gaps)** between them.
- Suitable for comparing categories.

---

# 3. Line Chart

## Definition

A **Line Chart** is a graph used to display how a numerical value changes over an **ordered sequence**, most commonly time. Individual observations are connected using straight lines to emphasize continuity and trends.

---

## Purpose

A line chart helps answer questions such as:

- Is the value increasing or decreasing?
- What trend does the data follow?
- How does the value change over time?

---

## When to Use

Use a line chart when:

- The data is ordered.
- The x-axis represents time or another meaningful sequence.
- You want to visualize trends or continuous changes.

---

## Examples

### Example 1

Question:

> How did the temperature change during one week?

Correct Graph:

**Line Chart**

Reason:

The data is recorded over time.

---

### Example 2

Question:

> How did monthly sales change throughout the year?

Correct Graph:

**Line Chart**

Reason:

Months have a natural chronological order.

---

## Important Characteristics

- Best suited for **time-series data**.
- Points are connected by lines.
- Highlights trends and continuous changes.
- The order of observations is important.

---

# 4. Scatter Plot

## Definition

A **Scatter Plot** is a graph used to display the relationship between **two numerical variables**. Each observation is represented by a single point on the graph.

The position of each point is determined by its x-value and y-value.

---

## Purpose

A scatter plot helps answer questions such as:

- Are two variables related?
- Is there a positive or negative correlation?
- Are there clusters in the data?
- Are there any outliers?

---

## When to Use

Use a scatter plot when:

- Both variables are numerical.
- You want to study the relationship between two variables.
- You want to detect trends, clusters, or outliers.

---

## Examples

### Example 1

Question:

> Is there a relationship between flipper length and bill length?

Correct Graph:

**Scatter Plot**

Reason:

Both variables are numerical, and the objective is to study their relationship.

---

### Example 2

Question:

> Is there a relationship between hours studied and exam score?

Correct Graph:

**Scatter Plot**

Reason:

The graph helps determine whether students who study more generally obtain higher scores.

---

## Important Characteristics

- Uses individual points.
- Every point represents one observation.
- Suitable for studying correlations.
- Useful for detecting clusters and outliers.

---

# 5. Box Plot

## Definition

A **Box Plot** is a graph used to summarize the distribution of numerical data using the **median, quartiles, interquartile range (IQR), and potential outliers**.

Instead of displaying every individual observation, it provides a compact summary of the data.

---

## Purpose

A box plot helps answer questions such as:

- Which group has the highest median?
- Which group has greater variability?
- Are there any outliers?
- How do the distributions differ between groups?

---

## When to Use

Use a box plot when:

- Comparing numerical data across different categories.
- Comparing distributions rather than individual values.
- Identifying variability and outliers.

---

## Examples

### Example 1

Question:

> Which penguin species is the heaviest on average, and how much do they vary?

Correct Graph:

**Box Plot**

Reason:

The graph compares the distribution of body mass for each species while also showing variation and outliers.

---

### Example 2

Question:

> Compare exam scores of students from three different classes.

Correct Graph:

**Box Plot**

Reason:

The graph summarizes the score distribution for each class and highlights differences in spread and median.

---

## Information Provided by a Box Plot

A box plot provides:

- Minimum value
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum value (excluding outliers)
- Interquartile Range (IQR)
- Potential outliers

---

## Important Characteristics

- Used for numerical data grouped by categories.
- Displays the median instead of the mean.
- Shows the spread of the data.
- Clearly identifies outliers.
- Useful for comparing multiple groups.

---

# 6. Pie Chart

## Definition

A **Pie Chart** is a circular graph divided into slices, where each slice represents the proportion or percentage of a category relative to the whole dataset.

The sum of all slices equals **100%**.

---

## Purpose

A pie chart helps answer questions such as:

- What percentage does each category contribute?
- How is the whole divided among different categories?

---

## When to Use

Use a pie chart when:

- Displaying proportions or percentages.
- Comparing a small number of categories.
- Showing how each category contributes to the total.

---

## Examples

### Example 1

Question:

> What percentage of students belong to each department?

Correct Graph:

**Pie Chart**

Reason:

The objective is to show the proportion of each department relative to the total number of students.

---

### Example 2

Question:

> What percentage of total company sales comes from each product category?

Correct Graph:

**Pie Chart**

Reason:

Each product contributes a proportion of the total sales.

---

## Important Characteristics

- Represents parts of a whole.
- Values are shown as percentages or proportions.
- Best for a small number of categories.
- Total of all slices equals 100%.

# Exam Tip

Before selecting a graph, always identify **what the question is asking**:

- **Distribution of numerical values** → Histogram
- **Comparison between categories** → Bar Chart
- **Trend or change over time** → Line Chart
- **Relationship between two numerical variables** → Scatter Plot
- **Comparison of distributions, median, spread, or outliers** → Box Plot
- **Percentage or proportion of a whole** → Pie Chart

Choosing the correct visualization depends on the **purpose of the analysis**, not simply on the type of graph.
