## Question

Explain why it is not necessary to manually define which specific filters (e.g., edge detection filters, sharpening filters) are used in the individual convolutional layers when training a CNN.

## Answer

In a Convolutional Neural Network (CNN), it is not necessary to manually define specific filters because the network learns the filter values automatically during training. The filters in a CNN are trainable parameters, similar to the weights and biases used in a Multi-Layer Perceptron (MLP).

At the beginning of training, the filter values are initialized with random numbers. During the training process, the CNN performs a forward pass to make predictions and calculates the prediction error using a loss function. Backpropagation and Gradient Descent are then used to update the filter values in a way that reduces the error. Over many training iterations, the network gradually learns filters that are useful for the given task.

For example, when training a CNN on the MNIST handwritten digit dataset, the network may automatically learn filters that detect edges, lines, corners, curves, and other patterns that help distinguish different digits. These filters are discovered directly from the training data without requiring manual design by the user.

This automatic feature learning is one of the major advantages of CNNs compared to traditional image processing methods. In traditional image processing, a human expert must manually design filters such as edge detectors or sharpening filters. In contrast, a CNN learns the most relevant filters for the specific problem automatically, often finding more effective feature representations than manually designed filters.

### Example

Traditional image processing:

- Human designs an edge detection filter.
- Human decides which filters should be applied.

CNN:

- Filters are initialized randomly.
- The network learns useful filters from data.
- Backpropagation updates filter values automatically.
- The learned filters become specialized for the target task.

### Conclusion

It is not necessary to manually define filters in a CNN because the filter values are learned automatically from the training data. Through backpropagation and gradient-based optimization, the network discovers the filters that best reduce the prediction error and improve performance for the given task.

## Question

Explain why CNN architectures usually consist of multiple convolutional layers.

## Answer

CNN architectures typically consist of multiple convolutional layers because different layers learn different levels of features from the input image. This allows the network to gradually build a more detailed and meaningful representation of the image.

The first convolutional layers usually learn simple features such as edges, lines, corners, and basic textures. These are often referred to as low-level features because they describe small local patterns in the image.

The outputs of the first layers are then used as inputs to deeper convolutional layers. These deeper layers combine simple features into more complex patterns such as curves, shapes, and object parts. These are known as mid-level features.

Even deeper layers learn high-level features by combining previously learned patterns into meaningful object representations. For example, when processing handwritten digits, deeper layers may learn complete digit structures rather than individual edges or curves.

This process is known as hierarchical feature learning, where simple features are gradually combined into increasingly complex representations. Multiple convolutional layers therefore allow the network to recognize complex visual patterns more effectively than a single convolutional layer.

Another advantage of using multiple convolutional layers is that the effective receptive field increases with depth. As a result, deeper layers can use information from larger regions of the image, helping the network understand both local details and global structures.

### Example

For handwritten digit recognition:

- Layer 1 may learn edges and lines.
- Layer 2 may learn curves and loops.
- Layer 3 may learn complete digit shapes.
- The final classification layer identifies the digit.

### Conclusion

CNNs usually contain multiple convolutional layers because they enable hierarchical feature learning. Early layers detect simple visual features, while deeper layers combine these features into more complex patterns and object representations. This improves the network's ability to recognize and classify images accurately.

### Question

Calculate the output feature matrix **S** of a convolution layer given the following input matrix **I**, kernel **K**, stride **s**, and padding **P**.

Input Matrix:

```text
1 2 1 0 2
2 1 0 3 1
3 1 2 1 0
0 2 1 3 1
1 0 2 1 2
```

Kernel:

```text
0  1  0
1 -4  1
0  1  0
```

Stride:

```text
s = 2
```

Padding:

```text
P = 1
```

---

### Step 1: Apply Padding

Since the padding value is 1, one layer of zeros is added around the input matrix.

Padded Input:

```text
0 0 0 0 0 0 0
0 1 2 1 0 2 0
0 2 1 0 3 1 0
0 3 1 2 1 0 0
0 0 2 1 3 1 0
0 1 0 2 1 2 0
0 0 0 0 0 0 0
```

---

### Step 2: Calculate Output Size

Output size is calculated using:

Output Size = ((N - F + 2P) / S) + 1

Where:

- N = 5
- F = 3
- P = 1
- S = 2

Calculation:

```text
((5 - 3 + 2×1) / 2) + 1
= (4 / 2) + 1
= 3
```

Therefore, the output feature map size is:

```text
3 × 3
```

---

### Step 3: Perform Convolution

After applying the kernel to each valid position and moving according to stride 2, the following output values are obtained:

```text
S(1,1) = 0
S(1,2) = -2
S(1,3) = -7

S(2,1) = -9
S(2,2) = -5
S(2,3) = 3

S(3,1) = -4
S(3,2) = -6
S(3,3) = -6
```

---

### Final Output Feature Matrix

```text
S =
[
  0  -2  -7
 -9  -5   3
 -4  -6  -6
]
```

---

### Explanation

Convolution is performed by sliding the kernel over the padded input image. At each position, corresponding values are multiplied and summed together to produce one output value. The stride determines how far the kernel moves after each calculation, while padding adds extra zeros around the image to preserve border information.

# Task 3: Convolutional Neural Network for MNIST Classification

## Question

Revisit the MNIST digits dataset, but this time train a Convolutional Neural Network (CNN) instead of a Multi-Layer Perceptron (MLP) for the classification task.

Summarize whether the CNN performed better than the previous MLP models and explain the reasons for the observed results.


````python

import tensorflow as tf
import tensorflow_datasets as tfds

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense
)

# Enable GPU memory growth if a GPU is available
gpus = tf.config.experimental.list_physical_devices("GPU")

for gpu in gpus:
    tf.config.experimental.set_memory_growth(gpu, True)


# Load the MNIST dataset
ds_train, ds_test = tfds.load(
    "mnist",
    split=["train", "test"],
    as_supervised=True,
    batch_size=32,
    shuffle_files=True
)


# Normalize pixel values to the range [0,1]
def normalize(image, label):

    image = tf.cast(image, tf.float32) / 255.0

    return image, label


ds_train = ds_train.map(normalize)
ds_test = ds_test.map(normalize)


# Build the CNN model
input_layer = Input(shape=(28, 28, 1))

# First convolution block
x = Conv2D(
    filters=32,
    kernel_size=(3, 3),
    activation="relu"
)(input_layer)

x = MaxPooling2D(
    pool_size=(2, 2)
)(x)

# Second convolution block
x = Conv2D(
    filters=64,
    kernel_size=(3, 3),
    activation="relu"
)(x)

x = MaxPooling2D(
    pool_size=(2, 2)
)(x)

# Convert feature maps into a vector
x = Flatten()(x)

# Fully connected layer
x = Dense(
    128,
    activation="relu"
)(x)

# Output layer for 10 digit classes
output_layer = Dense(
    10,
    activation="softmax"
)(x)

model = Model(
    inputs=input_layer,
    outputs=output_layer
)


# Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Display model structure and parameter count
model.summary()


# Train the model
history = model.fit(
    ds_train,
    epochs=10,
    verbose=1
)


# Evaluate on the test dataset
test_loss, test_accuracy = model.evaluate(
    ds_test,
    verbose=1
)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
````
