# Task 1: Conceptual Question


Q. Do you agree with the following statement?

“Unsupervised learning is not suitable for classification problems, but supervised learning and reinforcement learning are.”

- Give reasons for your answer
- Justify your explanation clearly

A.

I partially disagree with the statement.

Unsupervised learning is generally not suitable for classification problems because it does not use labeled data. In classification tasks, the model needs input-output pairs (for example, email → spam or not spam). Since unsupervised learning does not have such labels, it cannot directly perform classification.
Supervised learning is suitable for classification because the model is trained using labeled data. It learns the relationship between input and correct output, and can then predict the class of new data accurately.

However, reinforcement learning is not typically used for classification problems. Although it uses rewards and penalties, it is mainly designed for sequential decision-making tasks where the correct action is not known in advance. In classification, the correct answer is already available, so supervised learning is more appropriate.

Therefore, supervised learning is suitable for classification, unsupervised learning is not directly suitable, and reinforcement learning is generally not used for this purpose.


# Task 2: Choosing the Right Metric


For each of the following machine learning tasks:

- Select a suitable evaluation metric
- Explain why this metric is appropriate


1. Spam Emails: Classify emails as spam or not spam
   - False negatives are acceptable
   - False positives are annoying

   Precision is the most suitable evaluation metric for spam email classification because false positives (i.e., misclassifying legitimate emails as spam) are highly undesirable. Precision ensures that when the model predicts an email as spam, it is highly likely to be correct, thereby improving user trust and system reliability.


2. Medical Diagnosis: Predict whether a patient has a rare disease
   - Missing a positive case is very serious
   - False alarms are less critical

   Recall is the most suitable evaluation metric for medical diagnosis of a rare disease because false negatives (i.e., failing to detect an actual disease case) can have severe or life-threatening consequences. Recall ensures that the model identifies as many true positive cases as possible, minimizing missed diagnoses.


3. Rental Prices: Estimate monthly rent based on features
   - Large errors are not much worse than multiple small errors

   Mean Absolute Error (MAE) is the most suitable metric for rental price prediction because it treats all errors equally and provides a clear interpretation of the average prediction error. Since large errors are not significantly more critical than small errors in this context, MAE offers a simple and reliable evaluation measure.


4. Weather Prediction: Predict tomorrow’s average temperature
   - Both small and large errors matter
   - Large errors should be penalized more

   MSE/RMSE is the most suitable evaluation metric for weather prediction because it penalizes large errors more heavily than small ones. This is important since large deviations in temperature prediction can lead to significant real-world consequences. RMSE is especially useful as it provides error in the same unit as the predicted variable, making interpretation easier


5. Credit Card Fraud: Detect fraudulent transactions
   - Fraud cases are rare
   - Need to detect them reliably
   - But too many false alarms are costly

   F1 Score is the most suitable evaluation metric for credit card fraud detection because both false positives and false negatives are important. It provides a balance between precision and recall, ensuring that the model effectively detects fraudulent transactions while minimizing incorrect blocking of legitimate transactions, especially in highly imbalanced datasets.


# Task 3: Training vs Testing


- Explain what a training dataset is
- Explain what a test dataset is
- Why should a machine learning model not be evaluated on the same data it was trained on?

  The training dataset is the data used to teach a machine learning model, where the model learns patterns and relationships between inputs and outputs by adjusting its internal parameters. The test dataset is a separate set of data that the model has never seen before, and it is used after training to evaluate how well the model performs on new, unseen data. A model should not be evaluated on the same data it was trained on because it may simply memorize the training data instead of learning general patterns, leading to overly high and misleading performance results; this problem is known as overfitting. As a result, although the model may perform perfectly on training data, it may fail when applied to real-world data, which is why a separate test dataset is necessary to measure the model’s true ability to generalize.

## Task 6: Implementing a Guessing Model

Implement a simple model that ignores input features and only learns survival probabilities from the training data. It should have the functions `fit` and `predict`, that use the same parameters and serve a similar function as they did for the Decision Tree.

- The method `fit` should ignore the samples in `X_train` and simply learn the general probability of survival from the targets in `y_train`.
- The method `predict` should return array of predictions based on these learned probabilities for an input feature array.

Using your own model, call `fit` on `X_train` and `y_train` for training, then call `predict` on `X_test` to receive the predictions for the test data.


```python
import numpy as np

class GuessingModel:
    def fit(self, X, y):
        self.survival_prob = np.mean(y)

    def predict(self, X):
        predictions = np.random.choice(
            [0, 1],
            size=len(X),
            p=[1 - self.survival_prob, self.survival_prob]
        )
        return predictions


        
```

```python
gm = GuessingModel()

print("Training Guessing Model")
gm.fit(X_train, y_train)

print("Making predictions")
preds_gm = gm.predict(X_test)

print("First 10 predictions:")
print(preds_gm[:10])
```
