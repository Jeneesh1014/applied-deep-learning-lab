# Exercise: Choosing the Right Visualization

We are using the following dataset: `penguins` (available via `sns.load_dataset('penguins')`)

## Task 1: Distribution of Body Mass

**Question**: How is the body mass distributed across all penguins?
* **Option A**: A Line Plot (`sns.lineplot`) showing `body_mass_g` over the index.
* **Option B**: A Histogram (`sns.histplot`) of `body_mass_g`.

1. Create both plots.
2. Explain why `Option A` is misleading (Hint: Does the order of rows imply a time-series or a trend?).
3. Why is `Option B` the standard for this question?

## Task 2: Relationship between Flipper Length and Bill Length

**Question**: Is there a correlation between the length of a penguin's flippers and its bill?
* **Option A**: A Bar Plot (`sns.barplot`) showing the average `bill_length_mm` for every unique `flipper_length_mm`.
* **Option B**: A Scatter Plot (`sns.scatterplot`) with `flipper_length_mm` on the x-axis and `bill_length_mm` on the y-axis.

1. Create both plots.
2. Look at the Bar Plot. Why is it "overcrowded" and difficult to read?
3. How does the Scatter Plot help identify clusters or outliers that the Bar Plot hides?

## Task 3: Comparing Body Mass across Species

**Question**: Which penguin species is the heaviest on average, and how much do they vary?
* **Option A**: A Pie Chart (`Matplotlib plt.pie`) showing the sum of body mass per species.
* **Option B**: A Box Plot (`sns.boxplot`) showing body_mass_g per species.

1. Create both plots.
2. Why is the Pie Chart (Option A) a poor choice for comparing "heaviness"? (Hint: Does the sum of mass tell us anything if one species has more individuals than another?)
3. What information does the Box Plot provide that a simple average bar chart would miss? (Keywords: Median, Interquartile Range, Outliers).
