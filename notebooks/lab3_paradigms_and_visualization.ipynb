## Task 2: Relationship between Flipper Length and Bill Length

**Question**: Is there a correlation between the length of a penguin's flippers and its bill?

- **Option A**: A Bar Plot (`sns.barplot`) showing the average `bill_length_mm` for every unique `flipper_length_mm`.
- **Option B**: A Scatter Plot (`sns.scatterplot`) with `flipper_length_mm` on the x-axis and `bill_length_mm` on the y-axis.

2. Look at the Bar Plot. Why is it "overcrowded" and difficult to read?

```python
sns.barplot(data=penguins, x="flipper_length_mm", y="bill_length_mm")
plt.xticks(rotation=90)
plt.title("Bar plot flipper vs bill length")
plt.show()


# It is overcrowded because there are too many data points in the bar plot. also it is difficult to read and we cannot see the outliers either.
```

```python
sns.scatterplot(data=penguins, x="flipper_length_mm", y="bill_length_mm")
plt.title("Scatter plot flipper vs bill length")
plt.show()

# The scatter plot helps identify clusters or outliers that the bar plot hides. and also shows the relationship between this two numerical values.
```

## Task 4: Multi-dimensional Analysis (The "Expert" View)

**Question**: Does the relationship between flipper length and bill length differ by species?

- **Option A**: Three separate Scatter Plots in different windows (one for each species).
- **Option B**: One Scatter Plot using the hue parameter to color points by species.

1. Create both plots.
2. Explain the benefit of using hue.
3. Does the general trend across all penguins look different than the trend within a single species?


```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")
penguins = penguins.dropna(subset=["flipper_length_mm", "bill_length_mm", "species"])
```

```python
species_list = penguins["species"].unique()

for sp in species_list:
    subset = penguins[penguins["species"] == sp]
    
    sns.scatterplot(
        data=subset,
        x="flipper_length_mm",
        y="bill_length_mm"
    )
    
    plt.title(f"Scatter Plot for {sp}")
    plt.show()
```

```python
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="bill_length_mm",
    hue="species"
)

plt.title("Flipper vs Bill (Colored by Species)")
plt.show()
```
