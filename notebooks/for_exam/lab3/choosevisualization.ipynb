# Choosing the Correct Graph – Master Exam Notes

**Applied Deep Learning / Data Visualization**

> In university exams, the question is usually **not** "What is a histogram?" Instead, you are given a scenario and asked **which graph is the most appropriate**. The key is to understand **what information the question wants to show**.

---

# Step 1: Identify What the Question is Asking

Before selecting a graph, ask yourself these questions in order.

### Question 1

**Does the question ask about the distribution of a numerical variable?**

Examples:

- How are exam marks distributed?
- How is body mass distributed?
- What is the distribution of salaries?

→ **Use a Histogram**

---

### Question 2

**Does the question ask to compare different categories?**

Examples:

- Which species has the largest population?
- Compare sales of different products.
- Compare average salary of departments.

→ **Use a Bar Chart**

---

### Question 3

**Does the question ask about change over time?**

Examples:

- Monthly sales
- Daily temperature
- Stock prices
- Population growth over years

→ **Use a Line Chart**

---

### Question 4

**Does the question ask whether two numerical variables are related?**

Examples:

- Height vs Weight
- Flipper Length vs Bill Length
- Hours Studied vs Exam Score

→ **Use a Scatter Plot**

---

### Question 5

**Does the question ask to compare distributions between groups?**

Examples:

- Compare body mass across penguin species.
- Compare exam scores of different classes.
- Compare salaries across departments.

→ **Use a Box Plot**

---

### Question 6

**Does the question ask about percentages or proportions of a whole?**

Examples:

- Percentage of students in each department.
- Market share of different companies.
- Sales contribution by product category.

→ **Use a Pie Chart**

---

# One-Minute Exam Rule

If the question contains words like:

- **Distribution of numerical values** → Histogram
- **Compare categories / counts / averages** → Bar Chart
- **Trend / Over Time / Change** → Line Chart
- **Relationship / Correlation / Association** → Scatter Plot
- **Median / Spread / Variability / Outliers / Compare distributions** → Box Plot
- **Percentage / Share / Contribution / Proportion** → Pie Chart

Always choose the graph based on **what the question is asking**, not simply on the type of data.


# Tricky Visualization Exam Questions with Model Answers

**Applied Deep Learning – Master Exam Practice**

> These questions are designed in the style commonly asked in university exams. Many of them are intentionally tricky because multiple graphs may seem reasonable at first. Your task is to identify the **best** graph and explain **why**.

---

# Question 1

## Distribution of Body Mass

A zoologist wants to analyze how the **body mass of penguins is distributed** across the entire dataset.

Which graph is the most appropriate?

### Answer

**Histogram**

### Explanation

The question asks for the **distribution** of a numerical variable (body mass). A histogram groups body mass values into intervals (bins) and displays the frequency of observations in each interval, making it the standard graph for visualizing numerical distributions.

---

# Question 2

## Number of Penguins per Species

A researcher wants to compare how many penguins belong to each species.

Which graph is the most appropriate?

### Answer

**Bar Chart**

### Explanation

Species is a categorical variable. A bar chart compares the frequencies of different categories and clearly shows which species has the largest or smallest number of observations.

---

# Question 3 (Tricky)

## Average Body Mass per Species

A scientist wants to compare the **average body mass** of Adelie, Gentoo, and Chinstrap penguins.

Which graph is the most appropriate?

### Answer

**Bar Chart**

### Explanation

Although body mass is numerical, the comparison is between different categories (species). A bar chart is suitable because it compares aggregated numerical values (average body mass) across categorical groups.

---

# Question 4 (Very Tricky)

## Distribution of Body Mass per Species

A scientist wants to compare not only the average body mass but also the **spread, median, and outliers** of body mass for each penguin species.

Which graph is the most appropriate?

### Answer

**Box Plot**

### Explanation

The objective is to compare the distributions of body mass across species. A box plot summarizes each distribution using the median, quartiles, interquartile range (IQR), and potential outliers. A bar chart only shows averages and hides the variability within each species.

---

# Question 5

## Flipper Length and Bill Length

A biologist wants to determine whether penguins with longer flippers also tend to have longer bills.

Which graph is the most appropriate?

### Answer

**Scatter Plot**

### Explanation

Both flipper length and bill length are numerical variables. A scatter plot displays each penguin as an individual point, making it possible to identify correlations, trends, clusters, and outliers.

---

# Question 6

## Monthly Sales

A company records its total sales every month for one year and wants to visualize how sales changed over time.

Which graph is the most appropriate?

### Answer

**Line Chart**

### Explanation

The data is recorded in chronological order. A line chart emphasizes continuity and clearly illustrates trends, increases, decreases, and seasonal patterns over time.

---

# Question 7 (Common Trap)

## Country Population

A researcher wants to compare the population of Germany, France, Italy, and Spain.

Should a line chart be used?

### Answer

**No. A Bar Chart should be used.**

### Explanation

Countries are independent categories and have no natural sequential order. A line chart would incorrectly imply continuity between countries. A bar chart is the correct choice because it compares categorical values.

---

# Question 8 (Very Tricky)

## Student Marks

A teacher wants to know how student marks are distributed in an exam.

Which graph is the most appropriate?

### Answer

**Histogram**

### Explanation

Marks are numerical values, and the objective is to study their distribution. A histogram groups marks into intervals and shows the frequency of each range. A bar chart would not effectively represent the distribution of continuous numerical data.

# Final Memory Rule

Never choose a graph based only on the data type.

Always ask:

> **"What is the question trying to show?"**

- **Distribution** → Histogram
- **Comparison** → Bar Chart
- **Trend** → Line Chart
- **Relationship** → Scatter Plot
- **Distribution Comparison** → Box Plot
- **Proportion** → Pie Chart

If you answer this question first, you will select the correct visualization in almost every university exam.
