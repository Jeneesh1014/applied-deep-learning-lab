# Introduction

Data preprocessing is the process of converting raw data into a clean, consistent, and machine-readable format before training a machine learning model. Since real-world data often contains missing values, different scales, and categorical variables, preprocessing is essential to improve model performance and accuracy.

The main preprocessing techniques include:

1. Standardization
2. Normalization
3. Ordinal Encoding
4. One-Hot Encoding

---

# 1. Standardization

## Definition

Standardization is a feature scaling technique that transforms numerical data so that it has a **mean of 0** and a **standard deviation of 1**.

It ensures that all features have a similar scale while preserving the distribution of the data.

## Formula

[
x'=\frac{x-\mu}{\sigma}
]

Where:

- (x) = Original value
- (\mu) = Mean
- (\sigma) = Standard deviation

## Why is Standardization Needed?

Different features may have different scales.

Example:

| Feature | Value  |
| ------- | ------ |
| Age     | 25     |
| Salary  | 60,000 |

Without standardization, Salary dominates Age because of its larger numerical values.

Standardization brings both features to a comparable scale.

## Applications

- Logistic Regression
- Support Vector Machine (SVM)
- Neural Networks
- K-Means Clustering
- Principal Component Analysis (PCA)

## Advantages

- Prevents large-valued features from dominating.
- Improves convergence during training.
- Maintains the original distribution.

## Example

Original Data:

| Height (cm) |
| ----------- |
| 150         |
| 160         |
| 170         |
| 180         |

After Standardization:

| Standardized Value |
| ------------------ |
| -1.34              |
| -0.45              |
| 0.45               |
| 1.34               |

Mean becomes **0** and standard deviation becomes **1**.

# 2. Normalization (Min-Max Scaling)

## Definition

Normalization scales numerical values into a fixed range, usually **0 to 1**.

Unlike standardization, normalization changes the minimum and maximum values.

## Formula

[
x'=\frac{x-x_{min}}{x_{max}-x_{min}}
]

## Why is Normalization Needed?

Some machine learning algorithms require all input features to have the same range.

## Applications

- Neural Networks
- Image Processing
- Distance-based algorithms
- K-Nearest Neighbors (KNN)

## Advantages

- Converts all values to the same range.
- Faster model training.
- Prevents numerical instability.

## Example

Original Data:

| Marks |
| ----- |
| 20    |
| 40    |
| 60    |
| 80    |
| 100   |

After Normalization:

| Normalized Value |
| ---------------- |
| 0.00             |
| 0.25             |
| 0.50             |
| 0.75             |
| 1.00             |

---

# Difference Between Standardization and Normalization

| Standardization                          | Normalization                           |
| ---------------------------------------- | --------------------------------------- |
| Mean becomes 0                           | Values scaled between 0 and 1           |
| Standard deviation becomes 1             | Minimum becomes 0 and maximum becomes 1 |
| Preserves data distribution              | Changes data scale                      |
| Used when features have different scales | Used when fixed value range is required |

# 3. Ordinal Encoding

## Definition

Ordinal Encoding converts categorical variables into numerical values while **preserving the natural order** of the categories.

Each category is assigned an integer value according to its ranking.

## Example

Education Level

| Category  | Encoded Value |
| --------- | ------------- |
| Primary   | 0             |
| Secondary | 1             |
| Bachelor  | 2             |
| Master    | 3             |
| PhD       | 4             |

The numerical values represent increasing educational levels.

## Applications

- Education level
- Customer satisfaction
- Product quality
- Competition rankings

## Advantages

- Simple and efficient.
- Preserves meaningful order.
- Requires only one numerical column.

## Disadvantage

Should **only** be used when categories have a natural order.

---

# 4. One-Hot Encoding

## Definition

One-Hot Encoding converts categorical variables into multiple binary columns.

Each category is represented by a separate column containing either **0** or **1**.

Unlike Ordinal Encoding, no ordering is assumed.

## Example

Color Feature

Original Data

| Color |
| ----- |
| Red   |
| Blue  |
| Green |

After One-Hot Encoding

| Color_Red | Color_Blue | Color_Green |
| --------- | ---------- | ----------- |
| 1         | 0          | 0           |
| 0         | 1          | 0           |
| 0         | 0          | 1           |

Only one column contains **1**, while all others contain **0**.

## Applications

- Gender
- Country
- City
- Product category
- Animal species

## Advantages

- No false ordering between categories.
- Suitable for nominal categorical data.
- Widely used in machine learning.

## Disadvantages

- Increases the number of features.
- Large datasets with many categories produce many columns.

---

# Difference Between Ordinal Encoding and One-Hot Encoding

| Ordinal Encoding                | One-Hot Encoding                   |
| ------------------------------- | ---------------------------------- |
| Used when categories have order | Used when categories have no order |
| One numerical column            | Multiple binary columns            |
| Preserves ranking               | No ranking assumed                 |
| Example: Education Level        | Example: Color, Country            |

---

# Summary Table

| Technique        | Purpose                                          | Output                           |
| ---------------- | ------------------------------------------------ | -------------------------------- |
| Standardization  | Scale features using mean and standard deviation | Mean = 0, Standard Deviation = 1 |
| Normalization    | Scale values to a fixed range                    | Values between 0 and 1           |
| Ordinal Encoding | Convert ordered categories into integers         | Integer values                   |
| One-Hot Encoding | Convert unordered categories into binary vectors | Multiple binary columns          |

---

# Conclusion

Data preprocessing is an essential step in Machine Learning because it prepares raw data for model training. Standardization and Normalization scale numerical features, while Ordinal Encoding and One-Hot Encoding convert categorical variables into numerical form. Proper preprocessing improves model accuracy, training efficiency, and overall performance.

- **Standardization:** Mean = 0, Standard Deviation = 1.
- **Normalization:** Scale values between 0 and 1.
- **Ordinal Encoding:** Convert ordered categories into integers.
- **One-Hot Encoding:** Convert unordered categories into binary vectors.
