# Question 1. What is Seaborn? Explain its general purpose and one practical use case.

## Answer

**Seaborn** is an open-source Python library used for **statistical data visualization**. It is built **on top of Matplotlib**, meaning it uses Matplotlib as its underlying plotting engine while providing a simpler interface and more attractive default visualizations.

The main purpose of Seaborn is to create informative and visually appealing statistical graphics with minimal code. It works seamlessly with **Pandas DataFrames**, making it particularly useful for analyzing structured datasets.

Seaborn automatically applies better color palettes, themes, and statistical plotting features, reducing the amount of customization required compared to Matplotlib.

### Practical Use Case

A common use case of Seaborn is **visualizing survival rates in the Titanic dataset**. For example, a bar plot can compare the average survival rate of male and female passengers across different passenger classes, making patterns in the data easy to understand.

# Question 2. Why is Seaborn built on top of Matplotlib? Compare Seaborn and Matplotlib.

## Answer

Seaborn is built on top of Matplotlib because Matplotlib provides the fundamental plotting functionality, while Seaborn extends it by offering a simpler interface and improved statistical visualization capabilities.

Although both libraries are used for data visualization, they serve different purposes.

| Matplotlib                            | Seaborn                                      |
| ------------------------------------- | -------------------------------------------- |
| General-purpose plotting library      | Statistical visualization library            |
| More code is often required           | Simpler syntax with fewer lines of code      |
| Basic default appearance              | Attractive default themes and color palettes |
| Requires manual customization         | Automatic styling and formatting             |
| Works well with NumPy arrays          | Works naturally with Pandas DataFrames       |
| Greater flexibility for customization | Faster creation of statistical plots         |

### Conclusion

Matplotlib provides complete control over plots, while Seaborn focuses on making statistical visualization easier and more visually appealing.

# Question 3. Explain the common types of plots available in Seaborn.

## Answer

Seaborn provides many statistical visualization functions. Some of the most commonly used plots are:

### 1. Bar Plot

Compares numerical values across different categories.

**Applications:**

- Survival rate by passenger class
- Average salary by department
- Average marks by subject

---

### 2. Scatter Plot

Shows the relationship between two numerical variables.

**Applications:**

- Relationship between height and weight
- Relationship between age and income

---

### 3. Box Plot

Displays the distribution of data and helps identify outliers.

**Applications:**

- Detecting unusual values
- Comparing data distributions between groups

---

### 4. Histogram

Shows the frequency distribution of numerical values.

**Applications:**

- Age distribution
- Exam score distribution

---

### 5. Heatmap

Represents numerical values using colors.

**Applications:**

- Correlation matrices
- Missing value visualization
- Feature relationship analysis

These plots help users understand the structure, distribution, and relationships within a dataset.

# Question 4. What are the advantages of using Seaborn?

## Answer

Seaborn provides several advantages for statistical data visualization.

### Advantages

- Simple and easy-to-use syntax.
- Attractive default themes and color palettes.
- Direct integration with Pandas DataFrames.
- Built-in statistical plotting functions.
- Less code compared to Matplotlib.
- Automatic handling of categorical variables.
- Ideal for Exploratory Data Analysis (EDA).
- Supports visualizing relationships between multiple variables.

Because of these advantages, Seaborn is widely used in Data Science and Machine Learning for exploratory data analysis and statistical visualization.
