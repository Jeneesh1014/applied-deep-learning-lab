# 1. Code Comprehension

## Q1. What does the following program do?

```python
numbers = [1, -10, 0, -5, -1000, 100, 7]
maximum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num

print(maximum)
```

### Answer

The program finds the **maximum (largest) element** in a list.

It initializes the variable `maximum` with the first element of the list. Then it iterates through each element using a `for` loop. During each iteration, the current element is compared with the stored maximum value. If a larger value is found, the maximum variable is updated. After all elements have been checked, the largest value is printed.

---

## Q2. Why is the first element used instead of writing `maximum = 0`?

### Answer

The first element is used because the list may contain only negative numbers. If `maximum` were initialized to `0`, the algorithm would produce incorrect results when all elements are negative. Initializing with the first element guarantees that the maximum value always belongs to the list.

# 2. Dynamic Typing

## Q3. What property of Python is demonstrated by the following code?

```python
x = 132453
x = "Game of Thrones"
x = True
```

### Answer

The code demonstrates **dynamic typing**.

Python is a dynamically typed programming language, meaning that variables do not have fixed data types. A variable can store values of different data types during program execution without requiring explicit type declarations. The data type is determined automatically at runtime.

---

## Q4. What is the purpose of the `type()` function?

### Answer

The `type()` function returns the data type (class) of an object. It is commonly used to determine whether a variable contains an integer, string, Boolean, list, float, or another data type.

Example:

```python
type(10)      # int
type("Hello") # str
type(True)    # bool
```

# 3. List Slicing

## Q5. What is list slicing?

### Answer

List slicing is a technique used to extract a portion of a list by specifying a start index, stop index, and optional step size. The start index is included, while the stop index is excluded.

General syntax:

```python
list[start : stop : step]
```

---

## Q6. Explain the different slicing variants.

| Syntax                  | Description                                                                      |
| ----------------------- | -------------------------------------------------------------------------------- |
| `list[start:stop]`      | Returns elements from the start index (inclusive) to the stop index (exclusive). |
| `list[start:]`          | Returns all elements from the specified start index to the end of the list.      |
| `list[:stop]`           | Returns all elements from the beginning up to the stop index (exclusive).        |
| `list[:]`               | Returns a copy of the entire list.                                               |
| `list[start:stop:step]` | Returns elements using the specified step size.                                  |
| `list[::-1]`            | Returns the list in reverse order.                                               |

---

## Q7. Does slicing also work for strings?

### Answer

Yes.

Strings are sequence data types in Python, just like lists. Therefore, string slicing follows exactly the same syntax and behavior as list slicing.

Example:

```python
text = "Python"

text[1:4]   # "yth"
text[::-1]  # "nohtyP"
```

---

## Q8. State three important rules of Python slicing.

### Answer

1. The **start index is included**, while the **stop index is excluded**.
2. A **positive step** moves from left to right, whereas a **negative step** moves from right to left.
3. If the direction of the step does not match the start and stop indices, the result is an empty list.

# 4. Random Numbers

## Q9. What is the purpose of `random.randint(a, b)`?

### Answer

The function `random.randint(a, b)` generates and returns a random integer between `a` and `b`, including both boundary values.

Example:

Possible outputs:

# 5. Conditional Expressions

## Q10. What is Python's conditional (ternary) operator?

### Answer

Python provides a one-line conditional expression that evaluates one value if the condition is true and another value if the condition is false.

General syntax:

```python
value_if_true if condition else value_if_false
```

Example:

```python
result = n*n if n >= 5 else -n
```

It provides a shorter alternative to an `if-else` statement.

# 6. Functions

## Q11. What is a function?

### Answer

A function is a reusable block of code that performs a specific task. Functions improve code organization, reduce duplication, increase readability, and make programs easier to maintain.

---

## Q12. What is the difference between defining and calling a function?

### Answer

**Function Definition**

A function definition specifies the function name, parameters, and body using the `def` keyword.

**Function Call**

A function call executes the function by providing the required arguments.

---

## Q13. What is the purpose of the `return` statement?

### Answer

The `return` statement sends a value from a function back to the caller. Once a `return` statement is executed, the function terminates immediately.

---

# 7. Keyword Arguments

## Q14. What are keyword arguments?

### Answer

Keyword arguments are arguments passed to a function by specifying the parameter name together with its value.

Example:

```python
area(width=5, length=10)
```

---

## Q15. State two advantages of keyword arguments.

### Answer

1. Arguments can be passed in any order because they are matched by parameter name.
2. They improve code readability by clearly indicating which value belongs to which parameter.

---

# 8. Default Parameters

## Q16. What is a default parameter?

### Answer

A default parameter is a function parameter that already has a predefined value. If no argument is supplied during the function call, the default value is used automatically.

Example:

```python
def greet(name="Guest"):
    print(name)
```

# 9. Python Classes

## Q17. What is a class?

### Answer

A class is a blueprint or template used to create objects. It defines the attributes (data) and methods (functions) that all objects of that class will possess.

---

## Q18. What is an object?

### Answer

An object is an instance of a class. It represents a real-world entity with its own attributes and behavior.

---

## Q19. What is the purpose of the `__init__()` method?

### Answer

The `__init__()` method is a special constructor that is automatically executed whenever an object is created. It is used to initialize the attributes of the object.

---

## Q20. What is the purpose of the `__str__()` method?

### Answer

The `__str__()` method returns a human-readable string representation of an object. It is automatically called when the object is printed using the `print()` function.

# 10. Assertions

## Q21. What is the purpose of the `assert` statement?

### Answer

The `assert` statement is used to verify that a condition is true during program execution. If the condition evaluates to `False`, Python raises an `AssertionError`. Assertions are commonly used for debugging and validating program inputs.
