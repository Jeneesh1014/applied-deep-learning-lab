# Question (a)

## Question

**What is the main difference between Feedforward Neural Networks (FNNs) and Recurrent Neural Networks (RNNs) regarding their network structure?**

---

## Big Idea

The biggest difference between a **Feedforward Neural Network (FNN)** and a **Recurrent Neural Network (RNN)** is **memory**.

- A Feedforward Neural Network **does not remember previous inputs**.
- A Recurrent Neural Network **can remember previous inputs** using a **hidden state**.

This memory allows an RNN to process **sequential data** such as text, speech, and time-series data.

---

## Feedforward Neural Network (FNN)

A Feedforward Neural Network is the simplest type of neural network.

### Structure

```
Input Layer
      ↓
Hidden Layer
      ↓
Output Layer
```

### Characteristics

- Information flows only in one direction (forward).
- No loops or feedback connections.
- No memory.
- Every input is processed independently.
- Previous inputs do not affect future outputs.

### Example

Suppose the inputs are:

```
Image 1 → Output

Image 2 → Output

Image 3 → Output
```

The network processes each image independently.

It completely forgets Image 1 before processing Image 2.

---

## Recurrent Neural Network (RNN)

An RNN is a neural network with memory.

It contains **recurrent (feedback) connections** that allow information from previous time steps to be reused.

### Structure

```
          Previous Hidden State
                  │
                  ▼
Input ─────► Hidden ─────► Output
                  ▲
                  │
             Feedback Loop
```

### Characteristics

- Has recurrent (feedback) connections.
- Stores previous information in a hidden state.
- Can remember previous inputs.
- Suitable for sequential data.

---

## Why is Memory Important?

Many real-world problems involve sequences.

Examples:

- Sentences
- Speech
- Stock prices
- Weather prediction
- Sensor data

Example sentence:

```
I love deep learning.
```

To understand the word **learning**, the network must remember:

- I
- love
- deep

An FNN cannot naturally remember these previous words.

An RNN can.

---

## Key Difference

| Feedforward Neural Network         | Recurrent Neural Network  |
| ---------------------------------- | ------------------------- |
| No memory                          | Has memory                |
| No feedback connections            | Has recurrent connections |
| Processes each input independently | Uses previous information |
| Best for non-sequential data       | Best for sequential data  |

---

## Exam-Ready Answer

A **Feedforward Neural Network (FNN)** processes information only in the forward direction from the input layer to the output layer. It has no feedback connections and no memory, so every input is processed independently. A **Recurrent Neural Network (RNN)** contains recurrent (feedback) connections that allow information from previous time steps to be stored in a hidden state. This enables RNNs to remember previous inputs and process sequential data such as text, speech, and time-series data.

# Question (b)

## Question

**What role does the hidden state play in an RNN, and how does it enable sequential data processing?**

---

## Big Idea

The **hidden state** is the **memory** of an RNN.

It stores useful information from previous time steps and passes it to the next time step.

Without the hidden state, an RNN would behave like a normal feedforward network.

---

## What is the Hidden State?

The hidden state is an internal memory vector.

It stores information learned from previous inputs.

It is usually written as:

```
hₜ
```

where:

- **t** = current time step
- **hₜ** = hidden state at time step t

---

## How Does it Work?

At every time step, the RNN receives:

1. Current input (`xₜ`)
2. Previous hidden state (`hₜ₋₁`)

These are combined to produce a new hidden state (`hₜ`).

```
Current Input
       +
Previous Hidden State
       ↓
 New Hidden State
```

The new hidden state is then passed to the next time step.

---

## Example

Sentence:

```
I → love → deep → learning
```

Time Step 1

```
Input = I

Memory = "I"
```

Time Step 2

```
Input = love

Memory = "I love"
```

Time Step 3

```
Input = deep

Memory = "I love deep"
```

Time Step 4

```
Input = learning

Memory = "I love deep learning"
```

The hidden state allows the network to understand the complete sentence.

---

## Why is the Hidden State Important?

The hidden state allows an RNN to:

- Remember previous information
- Understand context
- Learn patterns over time
- Process sequences correctly

Without the hidden state:

```
Word 1 → forgotten

Word 2 → forgotten

Word 3 → forgotten
```

With the hidden state:

```
Word1

↓

Word1 + Word2

↓

Word1 + Word2 + Word3

↓

Word1 + Word2 + Word3 + Word4
```

---

## Applications

The hidden state is useful in:

- Language translation
- Speech recognition
- Chatbots
- Text generation
- Handwriting recognition
- Time-series forecasting

---

## Exam-Ready Answer

The **hidden state** acts as the **memory** of an RNN. It stores information from previous time steps and combines it with the current input to produce a new hidden state. This hidden state is passed to the next time step, allowing the network to remember previous information, understand context, and process sequential data such as text, speech, and time-series data.

# Question (c)

## Question

**What is the vanishing gradient problem in RNNs, and why does it occur?**

---

## Big Idea

The **vanishing gradient problem** is one of the biggest challenges when training RNNs.

During training, the gradients become extremely small.

As a result, the network cannot learn information from earlier time steps.

---

## Why Does It Happen?

RNNs are trained using:

**Backpropagation Through Time (BPTT)**

During BPTT, gradients are repeatedly multiplied while moving backward through many time steps.

Example:

```
0.5 × 0.5 × 0.5 × 0.5 × 0.5

= 0.03125
```

If multiplication continues,

```
0.5¹⁰

= 0.000976
```

The gradient becomes almost zero.

---

## What Happens?

When the gradient becomes very small:

- Earlier weights receive almost no updates.
- Learning becomes very slow.
- The network forgets long-term information.

---

## Example

Sentence:

```
I grew up in Germany ...
...
I speak German.
```

To understand the word **German**, the network should remember **Germany**.

However, because the gradient becomes extremely small, the earlier information is lost.

The RNN struggles to learn this long-term relationship.

---

## Why is it a Problem?

The vanishing gradient problem prevents an RNN from learning:

- Long-term dependencies
- Long sequences
- Relationships between distant words or events

---

## Solution

Modern RNN architectures reduce this problem using:

- **LSTM (Long Short-Term Memory)**
- **GRU (Gated Recurrent Unit)**

These models are designed to preserve information over long sequences.

---

## Exam-Ready Answer

The **vanishing gradient problem** occurs during the training of RNNs when gradients become extremely small as they are propagated backward through many time steps. This happens because gradients are repeatedly multiplied by values smaller than one during Backpropagation Through Time (BPTT). As a result, the weights of earlier time steps receive very small updates, making it difficult for the network to learn long-term dependencies. Modern architectures such as **LSTM** and **GRU** were developed to reduce this problem.

where $m$ = input size, $n$ = hidden size, $o$ = output size.

**Given:**

$$

$$
W_{hy} = \begin{bmatrix}2 & -1\\1 & 1\end{bmatrix}, \quad
b_y = \begin{bmatrix}0\\3\end{bmatrix}
$$

**Formulas:**
$$h_t = f(W_{xh}x_t + W_{hh}h_{t-1} + b_h), \qquad y_t = W_{hy}h_t + b_y$$

---

### Time step 1 ($t=1$)

**Step 1 — $W_{xh}x_1$:**

$$
\begin{bmatrix}2 & -1 & 3 & 1\\0 & 2 & -1 & -2\end{bmatrix}
\begin{bmatrix}1\\2\\0\\-1\end{bmatrix} =
\begin{bmatrix}(2)(1)+(-1)(2)+(3)(0)+(1)(-1)\\(0)(1)+(2)(2)+(-1)(0)+(-2)(-1)\end{bmatrix} =
\begin{bmatrix}-1\\6\end{bmatrix}

**Step 2 — $W_{hh}h_0$:** since $h_0 = [0,0]^T$, this is $[0,0]^T$.

$$
\begin{bmatrix}-1\\6\end{bmatrix} + \begin{bmatrix}0\\0\end{bmatrix} + \begin{bmatrix}1\\-2\end{bmatrix} = \begin{bmatrix}0\\4\end{bmatrix}
$$

**Step 4 — apply ReLU:**

$$
h_1 = \text{ReLU}\left(\begin{bmatrix}0\\4\end{bmatrix}\right) = \begin{bmatrix}0\\4\end{bmatrix}
$$

**Step 5 — compute $y_1 = W_{hy}h_1 + b_y$:**

$$
\begin{bmatrix}2 & -1\\1 & 1\end{bmatrix}\begin{bmatrix}0\\4\end{bmatrix} + \begin{bmatrix}0\\3\end{bmatrix} =
\begin{bmatrix}-4\\4\end{bmatrix} + \begin{bmatrix}0\\3\end{bmatrix} =
\begin{bmatrix}-4\\7\end{bmatrix}
$$

$$\boxed{y_1 = \begin{bmatrix}-4\\7\end{bmatrix}}$$

$$
\begin{bmatrix}2 & -1 & 3 & 1\\0 & 2 & -1 & -2\end{bmatrix}
\begin{bmatrix}0\\-2\\4\\1\end{bmatrix} =

# Task 4: Implementing an RNN for the MNIST Dataset

Let’s assume that we are implementing a document scanner. Thus, the input image is scanned line by line. We can interpret each line as one particular time step. Thus, we can train an RNN to predict the class of the MNIST digit as early as possible. For this purpose, we need to generate an output \(y_t\) for each time step \(t\), which is fed into classification layers (thus, at least a Softmax layer is needed). Setting a threshold for the Softmax score allows us to decide how **confident** the network needs to be to generate the output.

To solve this task, perform the following steps:

- Subdivide each image into a sequence of **28 vectors of length 28**.
- Create a new training and test set by combining **1 up to 28** of these vectors to form sequences of image rows. Each sequence must be associated with the corresponding label (i.e., the class to predict).
- Build a network consisting of:
  - A single recurrent layer.
  - One or more MLP (Dense) layers (your decision).
  - A Softmax output layer, which performs the final classification.
  - Use the **TimeDistributed** layer class so that Keras applies the same Dense layers to every time step.
- Fit (train) the model.
- Evaluate the model on the test dataset. Unlike previous exercises, set a **confidence threshold** so that the network returns the predicted class as soon as its confidence meets or exceeds the specified threshold.
- Use the **accuracy** metric during both training and testing.

## Questions

1. What is the **average time step** at which the RNN makes its prediction?
2. Measure both:
   - The classification performance (accuracy) of the RNN.
   - The average decision time step.
3. Analyze how these results change for different confidence thresholds.

````python
"""
Exercise 11 - Task 4: Implementing an "Anytime" RNN for the MNIST dataset
===========================================================================

Idea:
Each 28x28 MNIST image is treated as a sequence of 28 time steps, where each
time step is one row (a vector of length 28) of the image - like a document
scanner reading the image line by line.

A SimpleRNN with return_sequences=True produces one hidden state per row.
A TimeDistributed Dense + Softmax head is applied to EVERY time step, so the
network outputs a class-probability distribution after every single row it
has seen so far (many-to-many).

During evaluation, we stop "reading" an image as soon as the softmax
confidence (max probability) reaches a given threshold, and take that as the
network's decision. This lets us measure a trade-off between speed
(how many rows/time steps were needed) and accuracy.
"""

from datetime import datetime
import os
import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow.keras.layers import Input, SimpleRNN, Dense, TimeDistributed
from tensorflow.keras.models import Model


# ---------------------------------------------------------------------------
# GPU setup
# ---------------------------------------------------------------------------
gpus = tf.config.experimental.list_physical_devices('GPU')
for gpu in gpus:
    tf.config.experimental.set_memory_growth(gpu, True)


# ---------------------------------------------------------------------------
# Load MNIST dataset
# ---------------------------------------------------------------------------
ds_train, ds_val, ds_test = tfds.load(
    "mnist",
    split=["train[:80%]", "train[80%:]", "test"],
    as_supervised=True,
    shuffle_files=True
)

# Normalize pixel values from [0, 255] to [0.0, 1.0]
normalize = lambda x, y: (tf.cast(x, tf.float32) / 255.0, tf.cast(y, tf.int32))
ds_train = ds_train.map(normalize)
ds_val   = ds_val.map(normalize)
ds_test  = ds_test.map(normalize)


# ---------------------------------------------------------------------------
# Step 1: Reshape each image into a sequence of 28 rows (time steps) and
#         replicate the label across all 28 time steps (many-to-many)
# ---------------------------------------------------------------------------
def prepare_sequences(x, y):
    """
    1. Reshape the image dimensions: from (28, 28, 1) to a time-series of
       (28, 28)  -> 28 time steps, each a row-vector of length 28.
    2. Replicate the single label 'y' so that a label exists for each of the
       28 time steps (Many-to-Many approach), since the loss/metric is
       computed per time step but the true class never changes.
    """
    # Drop the channel dimension: (28, 28, 1) -> (28, 28)
    x_seq = tf.squeeze(x, axis=-1)

    # Repeat the scalar label 28 times -> shape (28,)
    y_seq = tf.repeat(y, repeats=28)

    return x_seq, y_seq


BATCH_SIZE = 64
train_sequence_ds = ds_train.map(prepare_sequences).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_sequence_ds   = ds_val.map(prepare_sequences).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test_sequence_ds  = ds_test.map(prepare_sequences).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)


# ---------------------------------------------------------------------------
# Step 2: Build the "anytime" RNN classifier
# ---------------------------------------------------------------------------
def build_anytime_rnn():
    """
    Builds an RNN that performs a classification after EACH individual
    time step (each scanned row), using TimeDistributed Dense layers so
    that the same classification head is applied at every time step.
    """
    # Input interface for (28 time steps, 28 pixels per row)
    inputs = Input(shape=(28, 28))

    # 1. Recurrent layer.
    #    CRITICAL: return_sequences=True is required for Many-to-Many,
    #    i.e. we need the hidden state / output at EVERY time step,
    #    not just the last one.
    rnn_out = SimpleRNN(128, return_sequences=True)(inputs)

    # 2. One (or more) MLP / Dense layers, applied identically to every
    #    time step via TimeDistributed.
    hidden = TimeDistributed(Dense(64, activation="relu"))(rnn_out)

    # 3. Softmax output layer with 10 classes, again applied at every
    #    time step via TimeDistributed.
    outputs = TimeDistributed(Dense(10, activation="softmax"))(hidden)

    model = Model(inputs=inputs, outputs=outputs, name="Anytime_MNIST_RNN")
    return model


model = build_anytime_rnn()
model.summary()

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n--- Training Started ---")
model.fit(train_sequence_ds, epochs=3, validation_data=val_sequence_ds)


# ---------------------------------------------------------------------------
# Step 3: Evaluate with an early-stopping confidence threshold
# ---------------------------------------------------------------------------
def evaluate_with_threshold(model, dataset, confidence_threshold=0.8):
    """
    Evaluates the model manually, batch by batch.
    For every image, we stop reading further rows as soon as the softmax
    confidence (max probability) at the current time step meets or exceeds
    'confidence_threshold'. That time step's prediction becomes the final
    decision for the image.

    Returns:
        accuracy  - overall classification accuracy using the "early" decisions
        avg_step  - average number of rows (1-28) that had to be scanned
                    before a decision was made
    """
    total_samples = 0
    correct_predictions = 0
    total_time_steps_taken = 0

    print(f"\nEvaluating with Confidence Threshold: {confidence_threshold}")

    for x_batch, y_batch in dataset:
        predictions = model.predict_on_batch(x_batch)          # shape: (B, 28, 10)
        true_labels = y_batch[:, 0].numpy()                    # all 28 rows share the same true label

        for i in range(len(predictions)):
            sample_preds = predictions[i]                      # (28, 10) - predictions for one image
            true_label = true_labels[i]

            # Default: if the threshold is never reached, fall back to the
            # very last row (t = 27) and its prediction.
            chosen_step = 27
            predicted_class = np.argmax(sample_preds[-1])

            # Walk through the time steps (rows) 0 .. 27 in order and stop
            # as soon as the network is "confident enough".
            for t in range(28):
                confidence_t = np.max(sample_preds[t])         # softmax max-probability at row t
                if confidence_t >= confidence_threshold:
                    chosen_step = t
                    predicted_class = np.argmax(sample_preds[t])
                    break

            total_samples += 1
            total_time_steps_taken += (chosen_step + 1)        # +1 for 1-based indexing (Row 1 to 28)
            if predicted_class == true_label:
                correct_predictions += 1

    avg_step = total_time_steps_taken / total_samples
    accuracy = correct_predictions / total_samples

    print(f"-> Accuracy: {accuracy:.4f}")
    print(f"-> Average row steps until decision: {avg_step:.2f} / 28")

    return accuracy, avg_step


# ---------------------------------------------------------------------------
# Step 4: Run the evaluation for several confidence thresholds
# ---------------------------------------------------------------------------
thresholds_to_test = [0.5, 0.7, 0.9, 0.98]
results = {}
for thresh in thresholds_to_test:
    acc, step = evaluate_with_threshold(model, test_sequence_ds, confidence_threshold=thresh)
    results[thresh] = (acc, step)

print("\n--- Summary ---")
print(f"{'Threshold':>10} | {'Accuracy':>10} | {'Avg. Rows Used':>15}")
for thresh, (acc, step) in results.items():
    print(f"{thresh:>10} | {acc:>10.4f} | {step:>15.2f}")


"""
Discussion of the questions
============================

1) What is the average time step at which the RNN makes its prediction?
   -> This is exactly the 'avg_step' value returned/printed above. Typically,
      for confidently-written digits (e.g. threshold ~0.7-0.9) the network
      already reaches the required confidence after roughly 8-15 rows
      (out of 28), i.e. it does not need to see the whole digit before it
      is sure about the class. Simple/round shapes (e.g. 0, 1) are usually
      decided earlier than digits that are visually similar to others
      (e.g. 4 vs 9, 3 vs 8), which need more rows before the softmax score
      is high enough.

2) Classification performance and average decision time step:
   -> Both accuracy and avg_step are reported together for every threshold
      in the printed summary table above.

3) How do results change with different confidence thresholds?
   -> Low threshold (e.g. 0.5): the network decides very early (small
      avg_step) because it only needs a moderate confidence, but accuracy
      is lower because early, ambiguous evidence can lead to wrong guesses.
   -> High threshold (e.g. 0.98): the network needs to be almost certain,
      so it waits longer (avg_step approaches 28) and, in turn, achieves
      higher accuracy - close to the accuracy obtained when using the full
      28-row sequence.
   -> This shows the classic speed/accuracy trade-off of "anytime"
      classifiers: increasing the confidence threshold increases both the
      average number of time steps needed and the final accuracy, and vice
      versa.
"""
````
