## 1. Global Metrics (Micro-Averaging)
Micro-averaging sums up the individual True Positives (TP), False Positives (FP), and False Negatives (FN) across all classes ($i \in \{F, U, W, A\}$).

### Calculation of Totals
* **Total TP:** $\sum TP_i = 30 + 15 + 8 + 25 = 78$
* **Total FP:** $\sum FP_i = 7 + 9 + 2 + 4 = 22$
* **Total FN:** $\sum FN_i = 5 + 5 + 2 + 10 = 22$

### Micro-Precision ($P_{micro}$)
Precision measures the ratio of correct positive predictions to the total predicted positives.
$$P_{micro} = \frac{\sum TP_i}{\sum TP_i + \sum FP_i} = \frac{78}{78 + 22} = \frac{78}{100} = 0.78$$

### Micro-Recall ($R_{micro}$)
Recall measures the ratio of correct positive predictions to the total actual positives.
$$R_{micro} = \frac{\sum TP_i}{\sum TP_i + \sum FN_i} = \frac{78}{78 + 22} = \frac{78}{100} = 0.78$$

### Micro-F1 Score ($F1_{micro}$)
The F1-Score is the harmonic mean of Precision and Recall. The harmonic mean is used instead of the arithmetic mean to penalize extreme values.
$$F1_{micro} = 2 \cdot \frac{P_{micro} \cdot R_{micro}}{P_{micro} + R_{micro}}$$

**Step-by-step substitution:**
1. Substitute the values: $2 \cdot \frac{0.78 \cdot 0.78}{0.78 + 0.78}$
2. Calculate the numerator: $2 \cdot 0.6084 = 1.2168$
3. Calculate the denominator: $0.78 + 0.78 = 1.56$
4. Final division: $1.2168 / 1.56 = 0.78$

> **Important Concept:** In single-label multi-class classification, every mistake is counted twice (once as a False Positive for the wrong class and once as a False Negative for the correct class). Therefore, $\sum FP = \sum FN$, which leads to $P_{micro} = R_{micro} = F1_{micro} = Accuracy$.

## 2. Macro-Averaged F1-Score
The Macro-F1 is the arithmetic mean of the individual F1-scores of each class. This treats all classes equally, regardless of their size.

### Step 1: Calculate individual F1-Scores
Using the formula $F1_i = 2 \cdot \frac{P_i \cdot R_i}{P_i + R_i}$

### Step 2: Calculate Macro-Average
Example for Exercise Sheet 4: $$F1_{macro} = \frac{F1_F + F1_U + F1_W + F1_A}{4}$$

## 3. Accuracy and Global Metrics

### 3.1. The Accuracy Formula
Accuracy is the most straightforward metric. It calculates the ratio of all correct predictions (True Positives for all classes) to the total number of cases.

**Formula:**
$$Accuracy = \frac{\sum TP_i}{\text{Total Number of Samples}}$$

**Calculation for your data:**
$$Accuracy = \frac{30 + 15 + 8 + 25}{100} = \frac{78}{100} = 0.78$$

### 3.2. Micro vs. Macro Accuracy?
In the world of machine learning, you will almost always see just one "Accuracy" value. Here is why:

#### Why there is no "Micro Accuracy"
Micro-averaging means aggregating all TPs, FPs, and FNs first. If you do this for accuracy, you simply get the standard Accuracy formula shown above. Therefore:
$$\text{Micro-Accuracy} = \text{Overall Accuracy}$$

#### What about "Macro Accuracy"?
Technically, you could calculate a **Macro-Accuracy** by computing the accuracy for each 2x2 binary matrix and averaging them. 
* **However, this is rarely used and often misleading.**
* In a 2x2 matrix, "Accuracy" includes **True Negatives (TN)**. In your Forest example, the TN is 58. A high TN count (all the things that are correctly identified as *not* being Forest) would inflate the accuracy score of that specific class. 
* Because Macro-averaging is intended to show how well the model performs on small classes, we use **Macro-F1** or **Macro-Precision/Recall** instead, as they ignore the "easy" True Negatives.

## 4. Accuracy: Binary vs. Multi-Class

### 4.1. Binary Accuracy (The 2x2 Perspective)
In a binary setup (e.g., "Is it a Forest? Yes/No"), this formula is correct:
$$Accuracy_{binary} = \frac{TP + TN}{TP + FP + FN + TN}$$
* Here, **TN (True Negatives)** are all the images that are correctly identified as "Not Forest".

### 4.2. Multi-Class Accuracy (The 4x4 Perspective)
When looking at the entire 100-image dataset across all four classes (Forest, Urban, Water, Agriculture), we define Accuracy as the sum of all correct predictions divided by the total number of samples.

**The Logic:**
Every "correct" prediction in a multi-class matrix sits on the **main diagonal**. For the Forest class, it's a TP. For the Urban class, its own correct predictions are also TPs. In the global view, we simply sum these up.

**Formula:**
$$Accuracy_{global} = \frac{\sum TP_i}{\text{Total Samples}}$$

**Calculation:**
$$Accuracy = \frac{TP_{Forest} + TP_{Urban} + TP_{Water} + TP_{Agriculture}}{100}$$
$$Accuracy = \frac{30 + 15 + 8 + 25}{100} = 0.78$$

### 4.3. Why we "leave out" TN in Multi-Class Accuracy
If you were to use the binary formula for the global model, you would run into a **double-counting problem**:

1.  **Redundancy:** A "True Negative" for the Forest class (e.g., a correctly identified Water image) is already counted as a "True Positive" for the Water class. 
2.  **Inflation:** If you summed up all TPs and all TNs from your four 2x2 matrices, you would get a number much higher than your total sample size (100), because each correct prediction would be counted multiple times (once as TP and three times as TN for the other classes).

### 4.4. Summary Table
| Context | Correct Predictions consist of... | Formula |
| :--- | :--- | :--- |
| **Binary (2x2)** | $TP + TN$ | $\frac{TP + TN}{Total}$ |
| **Multi-Class (4x4)** | All $TP_i$ (Diagonal) | $\frac{\sum TP_i}{Total}$ |

---

**Important Concept:** In Multi-Class metrics, we focus on the **Diagonal of the Confusion Matrix**. The Diagonal represents all instances where the `Actual Class == Predicted Class`.
