# Task

A machine learning model was trained to classify weather conditions into three categories: **Sunny** (S), **Cloudy** (C), and **Rainy** (R). After testing the model on 50 days, the following results were recorded:
* **Sunny (Actual):** 20 days total. 16 were correctly predicted as Sunny, 2 as Cloudy, and 2 as Rainy.
* **Cloudy (Actual):** 20 days total. 14 were correctly predicted as Cloudy, 4 a Sunny, and 2 as Rainy.
* **Rainy (Actual):** 10 days total. 8 were correctly predicted as Rainy and 2 as Cloudy.

**Do the following:**
* **Construct the Multi-Class Confusion Matrix:** Create a 3x3 matrix where rows represent the Actual Data and columns represent the Forecast (Predictions).
* **Binary Decomposition:** Create a 2x2 Confusion Matrix for each class.
* **Calculate Performance Metrics:** Macro-averaged Precision, Recall and the F1-Score for each of the Binary Decomposition 2x2 Confusion Matrices.

---

In this matrix, rows represent the **Actual** data and columns represent the **Forecast** (Predictions).

| **Cloudy (C)** | 4 | 14 | 2 | **20** |
| **Rainy (R)** | 0 | 2 | 8 | **10** |
| **Total** | **20** | **18** | **12** | **50** |

## 3. Performance Metrics

### Individual Class Metrics
The metrics are calculated based on the binary decomposition for each class.

| Class | Precision ($TP / (TP+FP)$) | Recall ($TP / (TP+FN)$) | F1-Score ($2 \cdot \frac{P \cdot R}{P+R}$) |
| :--- | :---: | :---: | :---: |
| **Sunny** | $\frac{16}{20}=\frac{4}{5}$ | $\frac{16}{20}=\frac{4}{5}$ | $\frac{16}{20}=\frac{4}{5}$ |
| **Cloudy** | $\frac{14}{18}=\frac{7}{9}$ | $\frac{14}{20}=\frac{7}{10}$ | $\frac{98}{133}=\frac{14}{19}$ |
| **Rainy** | $\frac{8}{12}=\frac{2}{3}$ | $\frac{8}{10}=\frac{4}{5}$ | $\frac{32}{44}=\frac{8}{11}$ |

### Macro-Averaged Metrics
The Macro-Average is the arithmetic mean of the individual class scores.

*   **Macro-Precision:** $\frac{1}{3} \cdot (\frac{16}{20} + \frac{14}{18} + \frac{8}{12}) = \frac{1}{3} \cdot (\frac{4}{5} + \frac{7}{9} + \frac{2}{3}) = \mathbf{\frac{101}{135}}$
*   **Macro-Recall:** $\frac{1}{3} \cdot (\frac{16}{20} + \frac{14}{20} + \frac{8}{10}) = \frac{1}{3} \cdot (\frac{4}{5} + \frac{7}{10} + \frac{4}{5}) = \mathbf{\frac{23}{30}}$
