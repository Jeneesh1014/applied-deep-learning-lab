```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import io


df = pd.read_csv('office_coffee.csv')

# Global Settings
sns.set_theme(style='darkgrid') # Options: 'white', 'dark', 'whitegrid', 'ticks'
```

```python
# --- 1. Histplot (1x Numeric) ---
# Purpose: Distribution/Spread of a single variable
# Parameters: 
#   bins: integer (e.g., 5, 10, 20)
#   kde: True/False (Kernel Density Estimate - the smooth line)
#   color: 'red', 'blue', 'green', etc.
plt.figure()
sns.histplot(data=df, x='Cups_per_Day', bins=5, kde=True, color='skyblue')
plt.title('Coffee Consumption Distribution')
plt.show()
```

```python
# --- 2. Barplot (1x Category + 1x Numeric) ---
# Purpose: Comparing means across groups
# Parameters:
#   palette: 'viridis', 'magma', 'rocket', 'deep'
#   estimator: 'mean', 'sum', 'median', 'max' (requires import numpy as np)
plt.figure()
sns.barplot(data=df, x='Department', y='Cups_per_Day', palette='viridis', hue='Department')
plt.title('Average Cups per Department')
plt.show()
```

```python
# --- 3. Boxplot (1x Category + 1x Numeric) ---
# Purpose: Statistical summary (Median, Quartiles, Outliers)
# Parameters:
#   width: float (e.g., 0.5, 0.8)
#   fliersize: float (size of outlier points)
plt.figure()
sns.boxplot(data=df, x='Remote', y='Satisfaction', width=0.5)
plt.title('Satisfaction: Remote vs. On-site')
plt.show()
```

```python
# --- 4. Violinplot (Density + Statistics) ---
# Purpose: Shows the 'shape' of the distribution data
# Parameters:
#   split: True (if you have two 'hue' categories, it splits the violin in half)
#   inner: 'box', 'quartile', 'point', 'stick', None
plt.figure()
sns.violinplot(data=df, x='Department', y='Satisfaction', inner='box', palette='muted', hue='Department')
plt.title('Satisfaction Density per Department')
plt.show()
```

```python
# --- 5. Heatmap (Correlation Matrix) ---
# Purpose: Visualize relationships between all numeric variables
# Parameters:
#   cmap: 'coolwarm', 'YlGnBu', 'Greens', 'Reds'
#   annot: True (shows values inside cells)
plt.figure()
correlation_matrix = df.corr(numeric_only=True)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
# --- 6. Pairplot (The 'Everything' Overview) ---
# Purpose: Plots every numeric variable against every other
# Parameters:
#   kind: 'scatter', 'kde', 'hist', 'reg' (regression line)
#   diag_kind: 'auto', 'hist', 'kde'
#   markers: ['o', 's', 'D'] (list of marker styles)
sns.pairplot(df, hue='Department', kind='scatter', diag_kind='kde', palette='Set2')
plt.show()
```

```python
# --- 7. Scatterplot (Numeric vs. Numeric) ---
# Purpose: Find out if more coffee leads to more satisfaction
# Parameters:
#   s: float -> Size of the dots (e.g., 100)
#   alpha: float -> Transparency (0.0 to 1.0) to see overlapping points
#   style: category -> Different marker shapes for a category (e.g., Remote)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df, 
    x='Cups_per_Day', 
    y='Satisfaction', 
    hue='Department',  # Color dots by Department
    style='Remote',    # Use different shapes (circles vs. crosses) for Remote status
    s=150,             # Increase point size
    alpha=0.8          # Slight transparency
)

plt.title("Relationship: Coffee vs. Satisfaction")
plt.xlabel("Cups per Day")
plt.ylabel("Satisfaction Score (1-10)")
plt.show()
```

```python
# Simple Regression Plot
sns.regplot(data=df, x='Cups_per_Day', y='Satisfaction')
plt.title("Trendline: Coffee impact on Satisfaction")
plt.show()
```
